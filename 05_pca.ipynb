<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Intuition &amp; Geometric Meaning</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">An <strong>eigenvector</strong> of a matrix A is a non-zero vector v that, when A is applied to it, only gets scaled — not rotated. The scaling factor is the <strong>eigenvalue</strong> λ. Formally: <strong>Av = λv</strong>. Geometrically, eigenvectors point along the "natural axes" of the linear transformation A. PCA uses this: the eigenvectors of the covariance matrix point in the directions of greatest variance.</p>
</div>

```python
# Geometric meaning: A transforms all vectors, but eigenvectors only get scaled
A = np.array([[2, 1],
              [1, 3]])

# Compute eigenvectors/eigenvalues
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Matrix A:")
print(A)
print(f"\nEigenvalue λ₁ = {eigenvalues[0]:.4f}, Eigenvector v₁ = {eigenvectors[:, 0]}")
print(f"Eigenvalue λ₂ = {eigenvalues[1]:.4f}, Eigenvector v₂ = {eigenvectors[:, 1]}")

# Verify Av = λv
for i in range(2):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    Av = A @ v
    lv = lam * v
    print(f"\nCheck: A @ v{i+1} = {np.round(Av, 4)}, λ{i+1} * v{i+1} = {np.round(lv, 4)}, Match: {np.allclose(Av, lv)}")
```

```python
# Visualize: eigenvectors are NOT rotated by A, only stretched
fig = go.Figure()

# Draw random vectors and their transformed versions
theta_vals = np.linspace(0, 2 * np.pi, 30, endpoint=False)
for t in theta_vals:
    v = np.array([np.cos(t), np.sin(t)])
    Av = A @ v
    fig.add_trace(go.Scatter(x=[0, v[0]], y=[0, v[1]], mode='lines',
        line=dict(color='#CCCCCC', width=1), showlegend=False))
    fig.add_trace(go.Scatter(x=[0, Av[0]], y=[0, Av[1]], mode='lines',
        line=dict(color='#E8B923', width=1), showlegend=False))

# Highlight eigenvectors
colors_ev = ['#4B2E83', '#C62828']
for i in range(2):
    v = eigenvectors[:, i]
    Av = A @ v
    fig.add_trace(go.Scatter(x=[0, v[0]], y=[0, v[1]], mode='lines+markers',
        line=dict(color=colors_ev[i], width=4), name=f'v{i+1} (eigenvector)'))
    fig.add_trace(go.Scatter(x=[0, Av[0]], y=[0, Av[1]], mode='lines+markers',
        line=dict(color=colors_ev[i], width=4, dash='dash'),
        name=f'Av{i+1} = {eigenvalues[i]:.2f}·v{i+1}'))

fig.update_layout(
    title='<b>Eigenvectors Stay on Their Line</b> — Only Scaled, Never Rotated',
    xaxis=dict(scaleanchor='y', range=[-4, 4]),
    yaxis=dict(range=[-4, 4]),
    template='plotly_white', height=500, width=550,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">2×2 Worked Example — Computing by Hand</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">For A = [[4, 2], [1, 3]], find eigenvalues by solving <strong>det(A − λI) = 0</strong>:<br>(4−λ)(3−λ) − 2·1 = 0 → λ² − 7λ + 10 = 0 → (λ−5)(λ−2) = 0.<br>Eigenvalues: <strong>λ₁ = 5, λ₂ = 2</strong>. Then substitute back to find each eigenvector.</p>
</div>

```python
# 2×2 eigenvalue computation by hand vs numpy
A2 = np.array([[4, 2],
               [1, 3]])

# By hand: det(A - λI) = (4-λ)(3-λ) - 2*1 = λ² - 7λ + 10 = 0
# Quadratic formula: λ = (7 ± √(49-40)) / 2 = (7 ± 3) / 2
lam1_hand = (7 + 3) / 2  # = 5
lam2_hand = (7 - 3) / 2  # = 2
print("=== By Hand ===")
print(f"Characteristic polynomial: λ² - 7λ + 10 = 0")
print(f"λ₁ = {lam1_hand}, λ₂ = {lam2_hand}")

# Eigenvector for λ₁=5: (A - 5I)v = 0 → [[-1, 2],[1, -2]]v = 0 → v₁ = [2, 1]
print(f"\nFor λ₁=5: (A-5I) = {A2 - 5*np.eye(2)}")
print("  -v₁₁ + 2v₁₂ = 0 → v₁ = [2, 1] (or any scalar multiple)")

# Eigenvector for λ₂=2: (A - 2I)v = 0 → [[2, 2],[1, 1]]v = 0 → v₂ = [-1, 1]
print(f"\nFor λ₂=2: (A-2I) = {A2 - 2*np.eye(2)}")
print("  2v₂₁ + 2v₂₂ = 0 → v₂ = [-1, 1] (or any scalar multiple)")

# Verify with numpy
print("\n=== NumPy Verification ===")
evals, evecs = np.linalg.eig(A2)
print(f"Eigenvalues: {evals}")
print(f"Eigenvectors (columns):\n{evecs}")
for i in range(2):
    v = evecs[:, i]
    print(f"Check Av{i+1} = λ{i+1}v{i+1}: {np.allclose(A2 @ v, evals[i] * v)}")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">3×3 Worked Example</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">For a 3×3 matrix the characteristic polynomial is cubic. We use numpy to solve but show the determinant setup. The key insight: a p×p matrix has p eigenvalues (counted with multiplicity) and up to p linearly independent eigenvectors.</p>
</div>

```python
# 3×3 eigenvalue example
A3 = np.array([[2, 0, 1],
               [0, 3, 0],
               [1, 0, 2]])

print("Matrix A (3×3):")
print(A3)

evals3, evecs3 = np.linalg.eig(A3)
print(f"\nEigenvalues: {np.round(evals3, 4)}")
print(f"\nEigenvectors (columns):")
print(np.round(evecs3, 4))

# Verify
for i in range(3):
    v = evecs3[:, i]
    lam = evals3[i]
    check = np.allclose(A3 @ v, lam * v)
    print(f"\nλ{i+1} = {lam:.4f}, v{i+1} = {np.round(v, 4)}, Av = λv? {check}")

print("\nNote: A is symmetric → eigenvectors are orthogonal.")
print(f"v₁ · v₂ = {np.dot(evecs3[:,0], evecs3[:,1]):.6f}")
print(f"v₁ · v₃ = {np.dot(evecs3[:,0], evecs3[:,2]):.6f}")
print(f"v₂ · v₃ = {np.dot(evecs3[:,1], evecs3[:,2]):.6f}")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Geometric Intuition — Max Variance Direction</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Imagine a cloud of 2D data points forming an elongated ellipse. PCA finds the line through the cloud along which the projected data has <strong>maximum variance</strong>. That line is PC1. PC2 is the direction orthogonal to PC1 with the next most variance. The key insight: <strong>the eigenvectors of the covariance matrix point exactly in these directions</strong>, and the eigenvalues tell you how much variance each direction captures.</p>
</div>

```python
# Generate correlated 2D data and show PCA directions
np.random.seed(42)
mean = [0, 0]
cov_true = [[3, 2], [2, 1.5]]
X = np.random.multivariate_normal(mean, cov_true, 200)

# Center the data
X_centered = X - X.mean(axis=0)

# Covariance matrix
cov_mat = np.cov(X_centered, rowvar=False)
print("Sample covariance matrix:")
print(np.round(cov_mat, 4))

# Eigen-decomposition
evals_pca, evecs_pca = np.linalg.eigh(cov_mat)
# Sort descending
idx_sort = np.argsort(evals_pca)[::-1]
evals_pca = evals_pca[idx_sort]
evecs_pca = evecs_pca[:, idx_sort]

print(f"\nPC1 direction: {np.round(evecs_pca[:, 0], 4)}, variance = {evals_pca[0]:.4f}")
print(f"PC2 direction: {np.round(evecs_pca[:, 1], 4)}, variance = {evals_pca[1]:.4f}")
```

```python
# Visualize: data cloud + eigenvectors as principal axes
fig = go.Figure()

fig.add_trace(go.Scatter(x=X_centered[:, 0], y=X_centered[:, 1],
    mode='markers', marker=dict(size=5, color='#4B2E83', opacity=0.5),
    name='Centered data'))

# Draw PC axes scaled by eigenvalue
colors_pc = ['#E8B923', '#C62828']
for i in range(2):
    v = evecs_pca[:, i] * np.sqrt(evals_pca[i]) * 2  # scale for visibility
    fig.add_trace(go.Scatter(
        x=[-v[0], v[0]], y=[-v[1], v[1]],
        mode='lines+markers',
        line=dict(color=colors_pc[i], width=4),
        marker=dict(size=10, symbol='arrow', angleref='previous'),
        name=f'PC{i+1} (λ={evals_pca[i]:.2f})'))

fig.update_layout(
    title='<b>PCA Directions</b> — Eigenvectors of the Covariance Matrix',
    xaxis=dict(scaleanchor='y', title='X₁'),
    yaxis=dict(title='X₂'),
    template='plotly_white', height=500, width=600,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Covariance Eigen-Decomposition</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">PCA algorithm: (1) Center the data (subtract column means). (2) Compute covariance matrix <strong>S = XᵀX / (n−1)</strong>. (3) Eigen-decompose: <strong>S = VΛVᵀ</strong> where V is the matrix of eigenvectors (loadings) and Λ is a diagonal matrix of eigenvalues. (4) Project: <strong>scores = X·V</strong>.</p>
</div>

```python
# PCA from scratch: center → covariance → eigen-decompose → project
def pca_from_scratch(X, n_components=None):
    """PCA from scratch using eigen-decomposition of covariance matrix."""
    # Step 1: Center
    mean = X.mean(axis=0)
    X_c = X - mean
    
    # Step 2: Covariance matrix
    n = X_c.shape[0]
    cov = (X_c.T @ X_c) / (n - 1)
    
    # Step 3: Eigen-decomposition
    eigenvalues, eigenvectors = np.linalg.eigh(cov)
    # Sort descending
    idx = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]
    
    if n_components:
        eigenvalues = eigenvalues[:n_components]
        eigenvectors = eigenvectors[:, :n_components]
    
    # Step 4: Project (scores)
    scores = X_c @ eigenvectors
    
    # Variance explained
    total_var = eigenvalues.sum() if n_components is None else np.sum(np.linalg.eigh(cov)[0])
    var_explained = eigenvalues / np.sum(np.linalg.eigh(cov)[0])
    
    return scores, eigenvectors, eigenvalues, var_explained, mean

# Test on our 2D data
scores, loadings, evals, var_exp, mu = pca_from_scratch(X)
print("PCA from scratch:")
print(f"Loadings (eigenvectors):\n{np.round(loadings, 4)}")
print(f"Eigenvalues: {np.round(evals, 4)}")
print(f"Variance explained: {np.round(var_exp * 100, 2)}%")
print(f"Scores shape: {scores.shape}")
print(f"First 5 scores:\n{np.round(scores[:5], 4)}")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Loadings &amp; Scores</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Loadings</strong> (eigenvectors) define the new coordinate axes — they tell you how each original variable contributes to each PC. <strong>Scores</strong> are the coordinates of each data point in the new PC space. Loadings are directions; scores are positions.</p>
</div>

```python
# Compare our from-scratch PCA with sklearn
from sklearn.decomposition import PCA

pca_sk = PCA(n_components=2)
scores_sk = pca_sk.fit_transform(X)

print("sklearn PCA:")
print(f"Components (loadings):\n{np.round(pca_sk.components_, 4)}")
print(f"Explained variance: {np.round(pca_sk.explained_variance_, 4)}")
print(f"Explained variance ratio: {np.round(pca_sk.explained_variance_ratio_ * 100, 2)}%")

# Note: sklearn stores loadings as ROWS of components_
# Signs may differ (eigenvectors are unique up to sign)
print(f"\nFirst 5 scores (sklearn):\n{np.round(scores_sk[:5], 4)}")
print(f"\nFirst 5 scores (ours):\n{np.round(scores[:5], 4)}")
print("\nNote: signs may be flipped — eigenvectors are unique only up to sign.")
```

The <strong>scree plot</strong> shows eigenvalues in decreasing order — look for the "elbow" where they level off.   <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Kaiser Rule</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The <strong>Kaiser rule</strong> (Kaiser-Guttman criterion): keep components with eigenvalue > 1 (when using the correlation matrix) or eigenvalue > average eigenvalue (when using the covariance matrix). Rationale: a component should explain at least as much variance as one original standardized variable.</p>
</div>

```python
# Kaiser rule on Iris (covariance-based)
avg_eval = np.mean(evals_iris)
print(f"Average eigenvalue: {avg_eval:.4f}")
print(f"\nKaiser rule (keep λ > {avg_eval:.4f}):")
for i, ev in enumerate(evals_iris):
    keep = '✓ KEEP' if ev > avg_eval else '✗ DROP'
    print(f"  PC{i+1}: λ = {ev:.4f} {keep}")

# Kaiser rule on correlation-based PCA
X_std = StandardScaler().fit_transform(X_iris)
pca_corr = PCA()
pca_corr.fit(X_std)
evals_corr = pca_corr.explained_variance_

print(f"\nCorrelation-based PCA (Kaiser: keep λ > 1.0):")
for i, ev in enumerate(evals_corr):
    keep = '✓ KEEP' if ev > 1.0 else '✗ DROP'
    print(f"  PC{i+1}: λ = {ev:.4f} {keep}")
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Biplots</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">A <strong>biplot</strong> superimposes two plots: (1) <strong>scores</strong> — each observation as a point in PC space, and (2) <strong>loadings</strong> — each original variable as an arrow. Arrow direction shows how the variable aligns with the PCs; arrow length shows how much that variable contributes. Variables that point in the same direction are positively correlated.</p>
</div>

```python
# Biplot: Iris dataset (scores + loading arrows)
pca2 = PCA(n_components=2)
scores_iris = pca2.fit_transform(X_iris)
loadings_iris = pca2.components_.T  # (4, 2): each row is a variable's loading

fig = go.Figure()

# Scores colored by species
colors_species = ['#4B2E83', '#E8B923', '#22D3A5']
for i, name in enumerate(target_names):
    mask = y_iris == i
    fig.add_trace(go.Scatter(
        x=scores_iris[mask, 0], y=scores_iris[mask, 1],
        mode='markers', name=name,
        marker=dict(size=7, color=colors_species[i], opacity=0.7)))

# Loading arrows (scaled for visibility)
scale = 3.0
for j, feat in enumerate(feature_names):
    fig.add_annotation(
        ax=0, ay=0, axref='x', ayref='y',
        x=loadings_iris[j, 0] * scale,
        y=loadings_iris[j, 1] * scale,
        showarrow=True, arrowhead=3, arrowsize=1.5,
        arrowwidth=2, arrowcolor='#C62828')
    fig.add_trace(go.Scatter(
        x=[loadings_iris[j, 0] * scale * 1.12],
        y=[loadings_iris[j, 1] * scale * 1.12],
        mode='text', text=[feat], textfont=dict(size=11, color='#C62828'),
        showlegend=False))

fig.update_layout(
    title='<b>Biplot</b> — Iris PCA (Scores + Loadings)',
    xaxis_title=f'PC1 ({pca2.explained_variance_ratio_[0]*100:.1f}%)',
    yaxis_title=f'PC2 ({pca2.explained_variance_ratio_[1]*100:.1f}%)',
    template='plotly_white', height=550, width=700,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">PCA via SVD</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">PCA can also be computed via the <strong>Singular Value Decomposition (SVD)</strong>: X = UΣVᵀ. The right singular vectors V are the loadings (same as eigenvectors of XᵀX). The scores are UΣ. The singular values σᵢ relate to eigenvalues by λᵢ = σᵢ²/(n−1). SVD is more numerically stable than forming the covariance matrix explicitly.</p>
</div>

```python
# PCA via SVD
X_c = X_iris - X_iris.mean(axis=0)
n = X_c.shape[0]

# SVD
U, S, Vt = np.linalg.svd(X_c, full_matrices=False)

# Loadings = rows of Vt (or columns of V)
loadings_svd = Vt.T  # (4, 4)

# Scores = U * S
scores_svd = U * S

# Eigenvalues from singular values
evals_svd = S**2 / (n - 1)
var_ratio_svd = evals_svd / evals_svd.sum()

print("PCA via SVD:")
print(f"Singular values: {np.round(S, 4)}")
print(f"Eigenvalues (from SVD): {np.round(evals_svd, 4)}")
print(f"Eigenvalues (sklearn):  {np.round(pca_iris.explained_variance_, 4)}")
print(f"\nVariance ratio (SVD):    {np.round(var_ratio_svd * 100, 2)}%")
print(f"Variance ratio (sklearn): {np.round(pca_iris.explained_variance_ratio_ * 100, 2)}%")
print(f"\nMatch: {np.allclose(evals_svd, pca_iris.explained_variance_, atol=1e-10)}")
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Correlation-Based PCA vs. Covariance-Based PCA</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">When variables have different units or scales, PCA on the <strong>covariance matrix</strong> is dominated by high-variance variables. Solution: <strong>standardize first</strong> (z-score), then PCA on the covariance of standardized data = PCA on the <strong>correlation matrix</strong>. Rule of thumb: if variables have different units or vastly different variances, use correlation-based PCA.</p>
</div>

```python
# Covariance vs. Correlation PCA comparison
# Create data with different scales
np.random.seed(42)
X_scale = np.column_stack([
    np.random.normal(100, 30, 200),   # Feature A: large scale
    np.random.normal(0.5, 0.1, 200),  # Feature B: small scale
    np.random.normal(50, 10, 200)     # Feature C: medium scale
])

# Covariance-based PCA
pca_cov = PCA()
pca_cov.fit(X_scale)

# Correlation-based PCA (standardize first)
X_std = StandardScaler().fit_transform(X_scale)
pca_cor = PCA()
pca_cor.fit(X_std)

fig = make_subplots(rows=1, cols=2,
    subplot_titles=['Covariance PCA (raw data)', 'Correlation PCA (standardized)'])

for i in range(3):
    fig.add_trace(go.Bar(x=[f'PC{i+1}'], y=[pca_cov.explained_variance_ratio_[i] * 100],
        marker_color=['#4B2E83', '#E8B923', '#22D3A5'][i],
        name=f'PC{i+1}', showlegend=(i < 3)), row=1, col=1)
    fig.add_trace(go.Bar(x=[f'PC{i+1}'], y=[pca_cor.explained_variance_ratio_[i] * 100],
        marker_color=['#4B2E83', '#E8B923', '#22D3A5'][i],
        showlegend=False), row=1, col=2)

fig.update_layout(template='plotly_white', height=400,
    title='<b>Covariance vs. Correlation PCA</b> — Effect of Standardization',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.update_yaxes(title_text='Variance Explained (%)', row=1, col=1)
fig.update_yaxes(title_text='Variance Explained (%)', row=1, col=2)
fig.show()

print("Covariance PCA loadings (PC1):")
print(f"  Feature A (scale~100): {pca_cov.components_[0, 0]:.4f}")
print(f"  Feature B (scale~0.5): {pca_cov.components_[0, 1]:.4f}")
print(f"  Feature C (scale~50):  {pca_cov.components_[0, 2]:.4f}")
print("\n-> PC1 is dominated by Feature A (largest variance)!")
print("\nCorrelation PCA loadings (PC1):")
print(f"  Feature A: {pca_cor.components_[0, 0]:.4f}")
print(f"  Feature B: {pca_cor.components_[0, 1]:.4f}")
print(f"  Feature C: {pca_cor.components_[0, 2]:.4f}")
print("\n-> Variables contribute more equally after standardization.")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">5×2 Data: Full PCA Pipeline</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">We walk through every step of PCA on a tiny 5×2 dataset: (1) Raw data, (2) Centering, (3) Covariance matrix, (4) Eigenvalues &amp; eigenvectors, (5) Loadings, (6) Scores, (7) Biplot. This makes every matrix operation visible.</p>
</div>

```python
# Step 0: Raw data (5 observations, 2 features)
X_raw = np.array([[2.5, 2.4],
                  [0.5, 0.7],
                  [2.2, 2.9],
                  [1.9, 2.2],
                  [3.1, 3.0]])

labels_5 = ['A', 'B', 'C', 'D', 'E']
print("Step 0: Raw Data (5×2)")
print(pd.DataFrame(X_raw, index=labels_5, columns=['X₁', 'X₂']))

# Step 1: Center
mean_5 = X_raw.mean(axis=0)
X_c5 = X_raw - mean_5
print(f"\nStep 1: Column means = {np.round(mean_5, 4)}")
print("Centered data:")
print(pd.DataFrame(np.round(X_c5, 4), index=labels_5, columns=['X₁', 'X₂']))

# Step 2: Covariance matrix
cov_5 = np.cov(X_c5, rowvar=False)
print(f"\nStep 2: Covariance matrix S = XᵀX / (n-1):")
print(np.round(cov_5, 4))

# Step 3: Eigenvalues and eigenvectors
evals_5, evecs_5 = np.linalg.eigh(cov_5)
idx5 = np.argsort(evals_5)[::-1]
evals_5 = evals_5[idx5]
evecs_5 = evecs_5[:, idx5]
print(f"\nStep 3: Eigenvalues: {np.round(evals_5, 4)}")
print(f"Eigenvectors (loadings):\n{np.round(evecs_5, 4)}")

# Step 4: Variance explained
var_exp_5 = evals_5 / evals_5.sum()
print(f"\nStep 4: Variance explained: {np.round(var_exp_5 * 100, 2)}%")

# Step 5: Scores (project centered data onto PCs)
scores_5 = X_c5 @ evecs_5
print(f"\nStep 5: Scores (projections):")
print(pd.DataFrame(np.round(scores_5, 4), index=labels_5, columns=['PC1', 'PC2']))
```

```python
# Step 6: Biplot for the 5×2 data
fig = go.Figure()

# Scores as points
fig.add_trace(go.Scatter(
    x=scores_5[:, 0], y=scores_5[:, 1],
    mode='markers+text', text=labels_5,
    textposition='top center',
    marker=dict(size=12, color='#4B2E83'),
    name='Scores'))

# Loading arrows
scale_5 = 2.0
feat_names_5 = ['X₁', 'X₂']
for j in range(2):
    fig.add_annotation(
        ax=0, ay=0, axref='x', ayref='y',
        x=evecs_5[j, 0] * scale_5, y=evecs_5[j, 1] * scale_5,
        showarrow=True, arrowhead=3, arrowsize=1.5,
        arrowwidth=2.5, arrowcolor='#C62828')
    fig.add_trace(go.Scatter(
        x=[evecs_5[j, 0] * scale_5 * 1.15],
        y=[evecs_5[j, 1] * scale_5 * 1.15],
        mode='text', text=[feat_names_5[j]],
        textfont=dict(size=13, color='#C62828'),
        showlegend=False))

fig.update_layout(
    title=f'<b>Biplot (5×2 Data)</b> — PC1: {var_exp_5[0]*100:.1f}%, PC2: {var_exp_5[1]*100:.1f}%',
    xaxis_title='PC1', yaxis_title='PC2',
    xaxis=dict(scaleanchor='y'),
    template='plotly_white', height=500, width=550,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">SVD Approach on the Same 5×2 Data</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">We repeat the PCA using SVD on the centered data to show that both approaches yield the same results.</p>
</div>

```python
# SVD approach on the 5×2 data
U5, S5, Vt5 = np.linalg.svd(X_c5, full_matrices=False)

print("SVD decomposition of centered data:")
print(f"U (5×2):\n{np.round(U5, 4)}")
print(f"\nSingular values: {np.round(S5, 4)}")
print(f"\nVᵀ (2×2):\n{np.round(Vt5, 4)}")

# Loadings from SVD = columns of V = rows of Vt transposed
loadings_svd5 = Vt5.T
print(f"\nLoadings (from SVD):\n{np.round(loadings_svd5, 4)}")
print(f"Loadings (from eigen):\n{np.round(evecs_5, 4)}")

# Scores from SVD = U * S
scores_svd5 = U5 * S5
print(f"\nScores (SVD):")
print(pd.DataFrame(np.round(scores_svd5, 4), index=labels_5, columns=['PC1', 'PC2']))
print(f"\nScores (eigen):")
print(pd.DataFrame(np.round(scores_5, 4), index=labels_5, columns=['PC1', 'PC2']))

# Eigenvalues from singular values
evals_from_svd = S5**2 / (5 - 1)
print(f"\nEigenvalues (from SVD): {np.round(evals_from_svd, 4)}")
print(f"Eigenvalues (from eigen): {np.round(evals_5, 4)}")
print(f"Match: {np.allclose(np.abs(scores_svd5), np.abs(scores_5))}")
```

  <h3 style="color: #F57C00; margin-top: 0;">Exercise 4.1: Add a Third Variable</h3>

```python
# ============================================================
#  PCA EXPLORER DASHBOARD
# ============================================================
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display
import numpy as np
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

# ── Theme ──
C = {
    "bg": "#0F1117", "card": "#171A23", "grid": "#262B3A",
    "text": "#E8EAF0", "muted": "#8B90A0",
    "primary": "#4B2E83", "accent": "#E8B923",
    "green": "#22D3A5", "pink": "#E8608A",
    "palette": ["#4B2E83","#E8B923","#22D3A5","#E8608A","#4F8EF7","#F57C00"]
}
BASE = dict(paper_bgcolor=C["card"], plot_bgcolor=C["card"],
            font=dict(family="Inter, Segoe UI, sans-serif", color=C["text"], size=12),
            margin=dict(l=55, r=30, t=60, b=45))

def style(fig):
    fig.update_xaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    fig.update_yaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    return fig

figs = {}

# ── Data setup ──
iris = load_iris()
X_iris = iris.data
y_iris = iris.target
feat_names = iris.feature_names
tgt_names = iris.target_names

pca_full = PCA()
scores_full = pca_full.fit_transform(X_iris)
loadings_full = pca_full.components_.T
evals_full = pca_full.explained_variance_
var_ratio_full = pca_full.explained_variance_ratio_
cum_var_full = np.cumsum(var_ratio_full)

# ============================================================
# TAB 1: Eigen — eigenvector visualization on 2D data
# ============================================================
mean_2d = [0, 0]
cov_2d = [[3, 2], [2, 1.5]]
X_2d = np.random.multivariate_normal(mean_2d, cov_2d, 200)
X_2d_c = X_2d - X_2d.mean(axis=0)
cov_2d_mat = np.cov(X_2d_c, rowvar=False)
evals_2d, evecs_2d = np.linalg.eigh(cov_2d_mat)
idx_2d = np.argsort(evals_2d)[::-1]
evals_2d = evals_2d[idx_2d]
evecs_2d = evecs_2d[:, idx_2d]

f = go.Figure()
f.add_trace(go.Scatter(x=X_2d_c[:, 0], y=X_2d_c[:, 1],
    mode='markers', marker=dict(size=4, color=C['muted'], opacity=0.5),
    name='Data points'))
pc_colors = [C['accent'], C['pink']]
for i in range(2):
    v = evecs_2d[:, i] * np.sqrt(evals_2d[i]) * 2.5
    f.add_trace(go.Scatter(
        x=[-v[0], 0, v[0]], y=[-v[1], 0, v[1]],
        mode='lines+markers',
        line=dict(color=pc_colors[i], width=4),
        marker=dict(size=[8, 0, 10]),
        name=f'PC{i+1} (λ={evals_2d[i]:.3f})'))
f.update_layout(**BASE, height=480,
    title=dict(text="<b>Eigenvectors of 2D Covariance Matrix</b>", font=dict(size=16)),
    xaxis=dict(scaleanchor='y', title='X₁'), yaxis=dict(title='X₂'))
style(f)
figs["eigen"] = f

# ============================================================
# TAB 2: PCA — biplot + scree
# ============================================================
# Biplot
f_bi = go.Figure()
species_colors = [C['primary'], C['accent'], C['green']]
for i, name in enumerate(tgt_names):
    mask = y_iris == i
    f_bi.add_trace(go.Scatter(
        x=scores_full[mask, 0], y=scores_full[mask, 1],
        mode='markers', name=name,
        marker=dict(size=6, color=species_colors[i], opacity=0.7)))
scale_bi = 3.5
for j, feat in enumerate(feat_names):
    f_bi.add_annotation(
        ax=0, ay=0, axref='x', ayref='y',
        x=loadings_full[j, 0] * scale_bi, y=loadings_full[j, 1] * scale_bi,
        showarrow=True, arrowhead=3, arrowsize=1.2,
        arrowwidth=2, arrowcolor=C['pink'])
    f_bi.add_trace(go.Scatter(
        x=[loadings_full[j, 0] * scale_bi * 1.15],
        y=[loadings_full[j, 1] * scale_bi * 1.15],
        mode='text', text=[feat.replace(' (cm)', '')],
        textfont=dict(size=11, color=C['pink']),
        showlegend=False))
f_bi.update_layout(**BASE, height=500,
    title=dict(text=f"<b>Iris Biplot</b> — PC1: {var_ratio_full[0]*100:.1f}%, PC2: {var_ratio_full[1]*100:.1f}%", font=dict(size=16)),
    xaxis=dict(title='PC1'), yaxis=dict(title='PC2'))
style(f_bi)
figs["biplot"] = f_bi

# Scree
f_sc = go.Figure()
f_sc.add_trace(go.Bar(
    x=[f'PC{i+1}' for i in range(4)], y=var_ratio_full * 100,
    marker_color=C['primary'], name='Individual'))
f_sc.add_trace(go.Scatter(
    x=[f'PC{i+1}' for i in range(4)], y=cum_var_full * 100,
    mode='lines+markers', marker=dict(size=10, color=C['accent']),
    line=dict(color=C['accent'], width=3), name='Cumulative', yaxis='y2'))
f_sc.add_hline(y=95, line_dash='dash', line_color=C['pink'],
    annotation=dict(text='95%', font=dict(color=C['pink'])))
f_sc.update_layout(**BASE, height=400,
    title=dict(text="<b>Scree Plot</b> — Iris PCA", font=dict(size=16)),
    yaxis=dict(title='Variance (%)'),
    yaxis2=dict(title='Cumulative (%)', overlaying='y', side='right',
        range=[0, 105], gridcolor=C['grid'], tickfont=dict(color=C['muted'])))
style(f_sc)
figs["scree"] = f_sc

# ============================================================
# TAB 3: Variance — cumulative explained + loading heatmap
# ============================================================
# Cumulative variance
f_cum = go.Figure()
f_cum.add_trace(go.Scatter(
    x=[f'PC{i+1}' for i in range(4)], y=cum_var_full * 100,
    mode='lines+markers+text',
    text=[f'{v*100:.1f}%' for v in cum_var_full],
    textposition='top center', textfont=dict(color=C['accent'], size=13),
    marker=dict(size=14, color=C['accent']),
    line=dict(color=C['accent'], width=3),
    fill='tozeroy', fillcolor='rgba(232,185,35,0.15)'))
f_cum.add_hline(y=95, line_dash='dash', line_color=C['pink'],
    annotation=dict(text='95% threshold', font=dict(color=C['pink'])))
f_cum.add_hline(y=80, line_dash='dot', line_color=C['green'],
    annotation=dict(text='80% threshold', font=dict(color=C['green'])))
f_cum.update_layout(**BASE, height=400,
    title=dict(text="<b>Cumulative Variance Explained</b>", font=dict(size=16)),
    yaxis=dict(title='Cumulative Variance (%)', range=[0, 110]))
style(f_cum)
figs["cumvar"] = f_cum

# Loading heatmap
loading_matrix = pca_full.components_  # (4, 4): rows are PCs
short_feat = [f.replace(' (cm)', '') for f in feat_names]
f_heat = go.Figure(go.Heatmap(
    z=loading_matrix,
    x=short_feat,
    y=[f'PC{i+1}' for i in range(4)],
    colorscale=[[0, C['pink']], [0.5, C['card']], [1, C['accent']]],
    zmid=0,
    text=np.round(loading_matrix, 3),
    texttemplate='%{text}',
    textfont=dict(size=12)))
f_heat.update_layout(**BASE, height=350,
    title=dict(text="<b>Loading Heatmap</b> — Variable Contributions to Each PC", font=dict(size=16)))
figs["heatmap"] = f_heat

# ============================================================
# Assemble tabbed dashboard
# ============================================================
def div(key, first=False):
    return figs[key].to_html(full_html=False,
        include_plotlyjs="cdn" if first else False,
        config={"displayModeBar": True, "displaylogo": False})

card = lambda inner: f"<div class='card'>{inner}</div>"
tabs = {
    "Eigen": card(div("eigen", first=True)),
    "PCA": card(div("biplot")) + card(div("scree")),
    "Variance": card(div("cumvar")) + card(div("heatmap")),
}
nav = "".join(
    f"<button class='tab-btn{' active' if i==0 else ''}' onclick=\"showTab(event,'tab{i}')\">"
    + name + "</button>" for i, name in enumerate(tabs))
panels = "".join(
    f"<div id='tab{i}' class='tab-panel' style='display:{'block' if i==0 else 'none'}'>{c}</div>"
    for i, c in enumerate(tabs.values()))

html = f"""<div id="dash"><style>
  #dash {{background:{C['bg']};border-radius:14px;padding:22px;font-family:Inter,'Segoe UI',sans-serif;}}
  #dash .dash-header {{display:flex;justify-content:space-between;align-items:baseline;margin-bottom:16px;}}
  #dash h1 {{color:{C['text']};font-size:21px;margin:0;}}
  #dash .sub {{color:{C['muted']};font-size:12px;}}
  #dash .tab-bar {{display:flex;gap:6px;border-bottom:1px solid {C['grid']};margin-bottom:18px;}}
  #dash .tab-btn {{background:none;border:none;color:{C['muted']};padding:10px 18px;font-size:13.5px;font-weight:600;cursor:pointer;border-bottom:2.5px solid transparent;}}
  #dash .tab-btn:hover {{color:{C['text']};}}
  #dash .tab-btn.active {{color:{C['accent']};border-bottom-color:{C['accent']};}}
  #dash .card {{background:{C['card']};border:1px solid {C['grid']};border-radius:12px;padding:6px;margin-bottom:16px;box-shadow:0 2px 10px rgba(0,0,0,.35);}}
</style>
<div class="dash-header"><h1>PCA Explorer</h1><span class="sub">BDATA 412 · Section 5</span></div>
<div class="tab-bar">{nav}</div>{panels}
<script>function showTab(evt,id){{const d=document.getElementById('dash');d.querySelectorAll('.tab-panel').forEach(p=>p.style.display='none');d.querySelectorAll('.tab-btn').forEach(b=>b.classList.remove('active'));document.getElementById(id).style.display='block';evt.currentTarget.classList.add('active');document.getElementById(id).querySelectorAll('.js-plotly-plot').forEach(g=>Plotly.Plots.resize(g));}}</script></div>"""
display(HTML(html))
print("Dashboard ready — switch tabs to explore Eigen, PCA, and Variance.")
```
