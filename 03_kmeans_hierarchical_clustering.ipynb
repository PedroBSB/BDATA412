  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">What is a Cluster?</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">A <strong>cluster</strong> is a group of data points that are more similar to each other than to points in other groups. Clustering is an <strong>unsupervised learning</strong> task — there are no labels. The algorithm must discover structure from the data alone. Key applications: customer segmentation, gene expression analysis, image compression, anomaly detection.</p>
</div>

```python
# Generate sample clustered data to illustrate the concept
np.random.seed(42)
X_demo, y_demo = make_blobs(n_samples=150, centers=3, cluster_std=0.8, random_state=42)

fig = go.Figure()
colors = ['#4B2E83', '#E8B923', '#22D3A5']
for k in range(3):
    mask = y_demo == k
    fig.add_trace(go.Scatter(
        x=X_demo[mask, 0], y=X_demo[mask, 1],
        mode='markers', name=f'Cluster {k+1}',
        marker=dict(size=9, color=colors[k], opacity=0.8)))

fig.update_layout(
    title='<b>Three Natural Clusters</b> — Can you see the groups?',
    xaxis_title='Feature X₁', yaxis_title='Feature X₂',
    template='plotly_white', height=450,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">The Data Matrix</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Given <strong>n</strong> observations each with <strong>p</strong> features, we arrange data into a matrix <strong>X ∈ ℝⁿˣᵖ</strong>. Each row <strong>xᵢ</strong> is one observation (a point in p-dimensional space). Each column is one feature.</p>
</div>

```python
# Data matrix example
data = np.array([
    [5.1, 3.5, 1.4],   # Obs 1
    [4.9, 3.0, 1.4],   # Obs 2
    [6.7, 3.1, 4.7],   # Obs 3
    [6.3, 2.5, 5.0],   # Obs 4
])
df_data = pd.DataFrame(data, columns=['Sepal Length', 'Sepal Width', 'Petal Length'],
                       index=['x₁', 'x₂', 'x₃', 'x₄'])
print("Data Matrix X ∈ ℝ⁴ˣ³:")
print(df_data)
print(f"\nShape: {data.shape} → n={data.shape[0]} observations, p={data.shape[1]} features")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Distance Matrix</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The <strong>distance matrix D ∈ ℝⁿˣⁿ</strong> stores pairwise distances: Dᵢⱼ = d(xᵢ, xⱼ). It is symmetric with zeros on the diagonal. Common distances: <strong>Euclidean</strong> d(x,y) = ‖x−y‖2, <strong>Manhattan</strong> d(x,y) = ‖x−y‖₁, <strong>Cosine distance</strong> = 1 − cos(θ).</p>
</div>

```python
# Distance matrix computation
from scipy.spatial.distance import pdist, squareform

D_condensed = pdist(data, metric='euclidean')
D = squareform(D_condensed)

labels = ['x₁', 'x₂', 'x₃', 'x₄']
df_dist = pd.DataFrame(np.round(D, 2), index=labels, columns=labels)
print("Distance Matrix D (Euclidean):")
print(df_dist)

# Visualize as heatmap
fig = go.Figure(go.Heatmap(
    z=D, x=labels, y=labels,
    colorscale=[[0, '#FFFFFF'], [0.5, '#4B2E83'], [1, '#32006E']],
    text=np.round(D, 2), texttemplate='%{text}',
    textfont=dict(size=14)))
fig.update_layout(
    title='<b>Pairwise Distance Matrix</b>',
    template='plotly_white', height=400, width=450,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Gram Matrix Trick</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The <strong>Gram matrix G = XXᵀ</strong> stores all pairwise dot products: Gᵢⱼ = xᵢ · xⱼ. The distance matrix can be recovered from G using: <strong>Dᵢⱼ² = Gᵢᵢ + Gⱼⱼ − 2Gᵢⱼ</strong>. This is the "kernel trick" foundation — you only need dot products to compute distances.</p>
</div>

```python
# Gram matrix and distance recovery
G = data @ data.T  # Gram matrix
print("Gram Matrix G = XX^T:")
print(pd.DataFrame(np.round(G, 2), index=labels, columns=labels))

# Recover distances from Gram matrix
diag = np.diag(G)
D_recovered = np.sqrt(diag[:, None] + diag[None, :] - 2 * G)
print("\nDistances recovered from Gram matrix:")
print(pd.DataFrame(np.round(D_recovered, 2), index=labels, columns=labels))
print(f"\nMatch original? {np.allclose(D, D_recovered)}")
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Similarity vs Dissimilarity</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Dissimilarity</strong> (distance): larger values = more different. Examples: Euclidean distance, Manhattan distance. <strong>Similarity</strong>: larger values = more alike. Examples: cosine similarity, correlation, RBF kernel k(x,y) = exp(−‖x−y‖² / 2σ²). You can convert: similarity = 1 − normalized_distance, or similarity = exp(−distance).</p>
</div>

```python
# Similarity vs Dissimilarity conversion
# Compute cosine similarity matrix
from sklearn.metrics.pairwise import cosine_similarity, rbf_kernel

cos_sim = cosine_similarity(data)
rbf_sim = rbf_kernel(data, gamma=0.1)

fig = make_subplots(rows=1, cols=3, subplot_titles=[
    'Euclidean Distance (Dissimilarity)', 'Cosine Similarity', 'RBF Kernel Similarity'])

for i, (mat, cscale) in enumerate([
    (D, [[0,'#FFFFFF'],[1,'#32006E']]),
    (cos_sim, [[0,'#32006E'],[1,'#E8B923']]),
    (rbf_sim, [[0,'#32006E'],[1,'#22D3A5']])
], 1):
    fig.add_trace(go.Heatmap(
        z=mat, x=labels, y=labels, colorscale=cscale,
        text=np.round(mat, 2), texttemplate='%{text}',
        textfont=dict(size=11), showscale=False), 1, i)

fig.update_layout(
    title='<b>Dissimilarity vs Similarity Measures</b>',
    template='plotly_white', height=380, width=950,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

  <h3 style="color: #F57C00; margin-top: 0;">Exercise 1.1: Build a Distance Matrix</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Given four 2D points A=(1,2), B=(4,6), C=(5,1), D=(2,3), compute the full 4×4 Euclidean distance matrix <strong>by hand</strong> (showing the formula), then verify with scipy. Which two points are closest? Which are farthest apart?</p>

```python
# Exercise 1.1: Your code here
points = np.array([[1,2], [4,6], [5,1], [2,3]])
point_labels = ['A', 'B', 'C', 'D']

# Manual computation
n = len(points)
D_manual = np.zeros((n, n))
for i in range(n):
    for j in range(n):
        D_manual[i, j] = np.sqrt(np.sum((points[i] - points[j])**2))
        if i < j:
            diff = points[i] - points[j]
            print(f"d({point_labels[i]},{point_labels[j]}) = sqrt({diff[0]}^2 + {diff[1]}^2) = {D_manual[i,j]:.3f}")

print("\nDistance Matrix:")
print(pd.DataFrame(np.round(D_manual, 3), index=point_labels, columns=point_labels))

# Verify with scipy
D_scipy = squareform(pdist(points))
print(f"\nMatch scipy? {np.allclose(D_manual, D_scipy)}")

# Find closest and farthest
mask = np.triu(np.ones((n,n), dtype=bool), k=1)
min_idx = np.argmin(D_manual[mask])
max_idx = np.argmax(D_manual[mask])
pairs = [(i,j) for i in range(n) for j in range(i+1,n)]
print(f"\nClosest: {point_labels[pairs[min_idx][0]]}-{point_labels[pairs[min_idx][1]]} (d={D_manual[pairs[min_idx][0], pairs[min_idx][1]]:.3f})")
print(f"Farthest: {point_labels[pairs[max_idx][0]]}-{point_labels[pairs[max_idx][1]]} (d={D_manual[pairs[max_idx][0], pairs[max_idx][1]]:.3f})")
```

  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Partition data into K groups by minimizing within-cluster variance</p>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">K-Means Objective Function</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">K-Means seeks to minimize the <strong>within-cluster sum of squares (WCSS)</strong>, also called <strong>inertia</strong>:</p>
  <p style="font-size: 1.15em; color: #333; text-align: center; margin: 15px 0;"><strong>J = Σᵏ Σ_{xᵢ ∈ Cₖ} ‖xᵢ − μₖ‖²</strong></p>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">where <strong>Cₖ</strong> is the set of points assigned to cluster k, and <strong>μₖ</strong> is the centroid (mean) of cluster k. This is an NP-hard optimization problem; Lloyd's algorithm provides an efficient heuristic.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Lloyd's Algorithm (K-Means)</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Step 0:</strong> Initialize K centroids (randomly or via k-means++).<br><strong>Step 1 (Assign):</strong> Assign each point to the nearest centroid: zᵢ = argminₖ ‖xᵢ − μₖ‖².<br><strong>Step 2 (Update):</strong> Recompute each centroid as the mean of its assigned points: μₖ = (1/|Cₖ|) Σ_{xᵢ ∈ Cₖ} xᵢ.<br><strong>Repeat</strong> Steps 1–2 until assignments stop changing or max iterations reached.<br><strong>Guarantee:</strong> J decreases (or stays equal) at every step → convergence to a local minimum.</p>
</div>

```python
# K-Means from scratch: Lloyd's Algorithm
def kmeans_lloyd(X, K, max_iter=100, seed=42):
    """Implement Lloyd's K-Means algorithm from scratch."""
    np.random.seed(seed)
    n, p = X.shape
    
    # Step 0: Random initialization
    idx = np.random.choice(n, K, replace=False)
    centroids = X[idx].copy()
    
    history = {'centroids': [centroids.copy()], 'labels': [], 'J': []}
    
    for iteration in range(max_iter):
        # Step 1: Assign each point to nearest centroid
        distances = np.zeros((n, K))
        for k in range(K):
            distances[:, k] = np.sum((X - centroids[k])**2, axis=1)
        labels = np.argmin(distances, axis=1)
        
        # Compute objective J (inertia)
        J = sum(distances[i, labels[i]] for i in range(n))
        history['labels'].append(labels.copy())
        history['J'].append(J)
        
        # Step 2: Update centroids
        new_centroids = np.zeros_like(centroids)
        for k in range(K):
            if np.sum(labels == k) > 0:
                new_centroids[k] = X[labels == k].mean(axis=0)
            else:
                new_centroids[k] = centroids[k]  # Keep if empty
        
        # Check convergence
        if np.allclose(centroids, new_centroids):
            print(f"Converged at iteration {iteration + 1}")
            break
        
        centroids = new_centroids
        history['centroids'].append(centroids.copy())
    
    return labels, centroids, history

# Run on demo data
X_blobs, y_true = make_blobs(n_samples=200, centers=3, cluster_std=1.0, random_state=42)
labels_scratch, centroids_scratch, hist = kmeans_lloyd(X_blobs, K=3)

print(f"Final inertia J = {hist['J'][-1]:.2f}")
print(f"Centroids:\n{np.round(centroids_scratch, 2)}")
```

```python
# Visualize Lloyd's algorithm step by step
n_steps = min(len(hist['centroids']), 6)
fig = make_subplots(rows=2, cols=3,
    subplot_titles=[f'Iteration {i}' if i > 0 else 'Initialization' for i in range(n_steps)])

colors_map = ['#4B2E83', '#E8B923', '#22D3A5']

for step in range(n_steps):
    row = step // 3 + 1
    col = step % 3 + 1
    
    if step == 0:
        # Show initial centroids with all points gray
        fig.add_trace(go.Scatter(
            x=X_blobs[:, 0], y=X_blobs[:, 1], mode='markers',
            marker=dict(size=5, color='#999'), showlegend=False), row, col)
    else:
        lab = hist['labels'][step-1]
        for k in range(3):
            mask = lab == k
            fig.add_trace(go.Scatter(
                x=X_blobs[mask, 0], y=X_blobs[mask, 1], mode='markers',
                marker=dict(size=5, color=colors_map[k], opacity=0.6),
                showlegend=False), row, col)
    
    # Plot centroids
    c = hist['centroids'][min(step, len(hist['centroids'])-1)]
    fig.add_trace(go.Scatter(
        x=c[:, 0], y=c[:, 1], mode='markers',
        marker=dict(size=15, color=['#4B2E83','#E8B923','#22D3A5'],
                    symbol='x', line=dict(width=2, color='black')),
        showlegend=False), row, col)

fig.update_layout(
    title="<b>Lloyd's Algorithm</b> — Step-by-Step Convergence",
    template='plotly_white', height=600, width=900,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">K-Means++ Initialization</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Random initialization can lead to poor local minima. <strong>K-means++</strong> (Arthur & Vassilvitskii, 2007) selects initial centroids that are spread apart:<br><strong>1.</strong> Choose first centroid uniformly at random.<br><strong>2.</strong> For each remaining centroid, pick point x with probability proportional to D(x)², its squared distance to the nearest existing centroid.<br><strong>3.</strong> Repeat until K centroids chosen.<br>This gives an O(log K)-competitive approximation to the optimal solution.</p>
</div>

```python
# K-Means++ initialization from scratch
def kmeans_plus_plus_init(X, K, seed=42):
    """K-means++ initialization."""
    np.random.seed(seed)
    n = X.shape[0]
    centroids = [X[np.random.randint(n)]]
    
    for _ in range(1, K):
        # Compute min squared distance to existing centroids
        dists = np.array([min(np.sum((x - c)**2) for c in centroids) for x in X])
        probs = dists / dists.sum()
        next_idx = np.random.choice(n, p=probs)
        centroids.append(X[next_idx])
    
    return np.array(centroids)

# Compare random vs k-means++ initialization
np.random.seed(42)
init_random = X_blobs[np.random.choice(len(X_blobs), 3, replace=False)]
init_pp = kmeans_plus_plus_init(X_blobs, 3)

fig = make_subplots(rows=1, cols=2, subplot_titles=['Random Init', 'K-Means++ Init'])
for i, (init, title) in enumerate([(init_random, 'Random'), (init_pp, 'K-Means++')], 1):
    fig.add_trace(go.Scatter(
        x=X_blobs[:, 0], y=X_blobs[:, 1], mode='markers',
        marker=dict(size=5, color='#999', opacity=0.4), showlegend=False), 1, i)
    fig.add_trace(go.Scatter(
        x=init[:, 0], y=init[:, 1], mode='markers',
        marker=dict(size=18, color=['#4B2E83','#E8B923','#22D3A5'],
                    symbol='star', line=dict(width=2, color='black')),
        name=title, showlegend=False), 1, i)

fig.update_layout(title='<b>Initialization Comparison</b> — Random vs K-Means++',
    template='plotly_white', height=400, width=800,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Matrix Form &amp; Assignment Matrix Z</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">In matrix form, define the <strong>assignment matrix Z ∈ {0,1}ⁿˣᵏ</strong> where Zᵢₖ = 1 if point i belongs to cluster k, else 0. Each row has exactly one 1 (hard assignment). The centroid matrix is <strong>M = (ZᵀZ)⁻¹ ZᵀX</strong>, and the objective becomes <strong>J = ‖X − ZM‖²_F</strong> (Frobenius norm squared).</p>
</div>

```python
# Assignment matrix Z demonstration
K = 3
n = len(X_blobs)
Z = np.zeros((n, K), dtype=int)
Z[np.arange(n), labels_scratch] = 1

print("Assignment Matrix Z (first 10 rows):")
print(pd.DataFrame(Z[:10], columns=[f'Cluster {k}' for k in range(K)]))
print(f"\nZ shape: {Z.shape}")
print(f"Each row sums to 1: {np.all(Z.sum(axis=1) == 1)}")
print(f"Cluster sizes: {Z.sum(axis=0)}")

# Recover centroids from matrix formula: M = (Z^T Z)^{-1} Z^T X
M = np.linalg.inv(Z.T @ Z) @ Z.T @ X_blobs
print(f"\nCentroids from matrix formula:\n{np.round(M, 2)}")
print(f"Match direct computation? {np.allclose(M, centroids_scratch)}")
```

  <h3 style="color: #F57C00; margin-top: 0;">Exercise 2.1: K-Means by Hand</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Given 6 points: (1,1), (1.5,2), (3,4), (5,7), (3.5,5), (4.5,5) and K=2 with initial centroids μ₁=(1,1) and μ₂=(5,7), perform <strong>two full iterations</strong> of Lloyd's algorithm. Show the assignment and update steps. Compute J at each iteration.</p>
</div>

```python
# Exercise 2.1: K-Means by hand
X_ex = np.array([[1,1], [1.5,2], [3,4], [5,7], [3.5,5], [4.5,5]])
centroids_ex = np.array([[1.0, 1.0], [5.0, 7.0]])
point_names = [f'P{i+1}' for i in range(6)]

for it in range(2):
    print(f"\n{'='*50}")
    print(f"ITERATION {it+1}")
    print(f"{'='*50}")
    print(f"Centroids: \u03bc₁={centroids_ex[0]}, \u03bc₂={centroids_ex[1]}")
    
    # Assignment step
    print("\n--- Assignment Step ---")
    labels_ex = []
    for i, x in enumerate(X_ex):
        d1 = np.sqrt(np.sum((x - centroids_ex[0])**2))
        d2 = np.sqrt(np.sum((x - centroids_ex[1])**2))
        assign = 0 if d1 <= d2 else 1
        labels_ex.append(assign)
        print(f"  {point_names[i]}={x}: d(μ₁)={d1:.2f}, d(μ₂)={d2:.2f} → Cluster {assign+1}")
    labels_ex = np.array(labels_ex)
    
    # Compute J
    J_ex = sum(np.sum((X_ex[i] - centroids_ex[labels_ex[i]])**2) for i in range(len(X_ex)))
    print(f"\nJ = {J_ex:.2f}")
    
    # Update step
    print("\n--- Update Step ---")
    for k in range(2):
        members = X_ex[labels_ex == k]
        centroids_ex[k] = members.mean(axis=0)
        print(f"  \u03bc{k+1} = mean({[point_names[i] for i in range(6) if labels_ex[i]==k]}) = {centroids_ex[k]}")
```

  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Build a tree of nested clusters from bottom up</p>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Agglomerative Hierarchical Clustering</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Bottom-up</strong> approach: start with each point as its own cluster, then repeatedly merge the two closest clusters until one cluster remains.<br><strong>Step 1:</strong> Compute pairwise distance matrix.<br><strong>Step 2:</strong> Find the two closest clusters.<br><strong>Step 3:</strong> Merge them into one cluster.<br><strong>Step 4:</strong> Update distances to the new cluster.<br><strong>Repeat</strong> until one cluster remains. The result is a <strong>dendrogram</strong> (tree diagram).</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Linkage Types</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The key question: how do you measure the distance between two <em>clusters</em>?</p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.8;">
    <li><strong>Single linkage:</strong> d(A,B) = min d(a,b) for a∈A, b∈B. Can produce "chaining" (elongated clusters).</li>
    <li><strong>Complete linkage:</strong> d(A,B) = max d(a,b). Produces compact, spherical clusters.</li>
    <li><strong>Average linkage:</strong> d(A,B) = mean of all pairwise distances. A compromise.</li>
    <li><strong>Ward's linkage:</strong> Merge the pair that minimizes the total within-cluster variance increase. Tends to produce equal-sized clusters.</li>
  </ul>
</div>

```python
# Generate data for hierarchical clustering
np.random.seed(42)
X_hc, y_hc = make_blobs(n_samples=50, centers=3, cluster_std=0.8, random_state=42)

# Compute linkage for all four methods
methods = ['single', 'complete', 'average', 'ward']
linkages = {}
for method in methods:
    linkages[method] = linkage(X_hc, method=method)
    print(f"{method.capitalize():>10s} linkage computed: {linkages[method].shape[0]} merges")
```

  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">A <strong>dendrogram</strong> is a tree diagram showing the sequence of merges. The y-axis shows the distance at which each merge occurs. To get K clusters, cut the dendrogram at a horizontal line — each branch below the cut is one cluster. Tall vertical lines indicate well-separated clusters.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Lance–Williams Formula</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">All four linkage methods can be expressed via the <strong>Lance–Williams recurrence</strong>. When clusters A and B merge into C = A ∪ B, the distance to any other cluster D is:<br><strong>d(C,D) = αₐ d(A,D) + αₕ d(B,D) + β d(A,B) + γ |d(A,D) − d(B,D)|</strong><br>Each linkage chooses different coefficients: Single (α=1/2, β=0, γ=−1/2), Complete (α=1/2, β=0, γ=+1/2), etc.</p>
</div>

```python
# Lance-Williams coefficients for each linkage type
lw_params = pd.DataFrame({
    'Linkage': ['Single', 'Complete', 'Average', 'Ward'],
    'α_A': ['1/2', '1/2', 'n_A/(n_A+n_B)', '(n_D+n_A)/(n_D+n_A+n_B)'],
    'α_B': ['1/2', '1/2', 'n_B/(n_A+n_B)', '(n_D+n_B)/(n_D+n_A+n_B)'],
    'β':   ['0', '0', '0', '-n_D/(n_D+n_A+n_B)'],
    'γ':   ['-1/2', '+1/2', '0', '0'],
    'Effect': ['min(d_AD, d_BD)', 'max(d_AD, d_BD)', 'weighted mean', 'minimize variance increase']
})
print("Lance\u2013Williams Coefficients:")
print(lw_params.to_string(index=False))
```

  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The <strong>cophenetic distance</strong> between two points is the height in the dendrogram where they first merge. The <strong>cophenetic correlation coefficient</strong> (CPCC) measures how well the dendrogram preserves pairwise distances: CPCC = corr(original distances, cophenetic distances). Values close to 1 indicate the dendrogram faithfully represents the data. Typically, <strong>average linkage</strong> yields the highest CPCC.</p>
</div>

```python
# Cophenetic correlation for each linkage method
original_dists = pdist(X_hc)

print("Cophenetic Correlation Coefficients:")
print("=" * 40)
cpcc_scores = {}
for method in methods:
    cpcc, coph_dists = cophenet(linkages[method], original_dists)
    cpcc_scores[method] = cpcc
    print(f"  {method.capitalize():>10s}: CPCC = {cpcc:.4f}")

best = max(cpcc_scores, key=cpcc_scores.get)
print(f"\nBest: {best.capitalize()} linkage (CPCC = {cpcc_scores[best]:.4f})")

# Visualize
fig = go.Figure(go.Bar(
    x=[m.capitalize() for m in methods],
    y=[cpcc_scores[m] for m in methods],
    marker_color=['#E8B923' if m == best else '#4B2E83' for m in methods],
    text=[f'{cpcc_scores[m]:.4f}' for m in methods],
    textposition='outside'))
fig.update_layout(
    title='<b>Cophenetic Correlation</b> — Which linkage preserves distances best?',
    xaxis_title='Linkage Method', yaxis_title='CPCC',
    yaxis_range=[0, 1.05],
    template='plotly_white', height=400,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Apply all four linkage methods to the Iris dataset (standardized). For each, cut the dendrogram at K=3 and compute the silhouette score. Which linkage best recovers the true species? Plot the dendrogram for the best method.</p>

```python
# Exercise 3.2: Linkage comparison on Iris
X_iris_std = StandardScaler().fit_transform(iris.data)

print("Linkage Comparison on Iris (K=3):")
print("=" * 55)
best_method = None
best_sil = -1

for method in methods:
    Z_iris = linkage(X_iris_std, method=method)
    labels_iris_hc = fcluster(Z_iris, t=3, criterion='maxclust')
    sil = silhouette_score(X_iris_std, labels_iris_hc)
    cpcc_val, _ = cophenet(Z_iris, pdist(X_iris_std))
    print(f"  {method.capitalize():>10s}: Silhouette={sil:.4f}, CPCC={cpcc_val:.4f}")
    if sil > best_sil:
        best_sil = sil
        best_method = method

print(f"\nBest linkage: {best_method.capitalize()} (Silhouette={best_sil:.4f})")

# Dendrogram for best method
Z_best = linkage(X_iris_std, method=best_method)
fig_iris, ax_iris = plt.subplots(figsize=(12, 5))
dendrogram(Z_best, ax=ax_iris, leaf_rotation=90, leaf_font_size=6,
           color_threshold=0.7 * max(Z_best[:, 2]),
           above_threshold_color='#999')
ax_iris.set_title(f'Iris Dendrogram ({best_method.capitalize()} Linkage)', fontsize=14,
                  fontweight='bold', color='#4B2E83')
ax_iris.set_xlabel('Sample Index')
ax_iris.set_ylabel('Distance')
ax_iris.axhline(y=0.7*max(Z_best[:, 2]), color='#E8B923', linestyle='--', label='Cut for K=3')
ax_iris.legend()
ax_iris.spines['top'].set_visible(False)
ax_iris.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
#  CLUSTERING EXPLORER DASHBOARD
# ============================================================
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, silhouette_samples
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import pdist

np.random.seed(42)

# ── Theme ──
C = {
    "bg": "#0F1117", "card": "#171A23", "grid": "#262B3A",
    "text": "#E8EAF0", "muted": "#8B90A0",
    "primary": "#4B2E83", "accent": "#E8B923",
    "green": "#22D3A5", "pink": "#E8608A",
    "palette": ["#4B2E83","#E8B923","#22D3A5","#E8608A","#4F8EF7","#F57C00"]
}
BASE = dict(paper_bgcolor=C["card"], plot_bgcolor=C["card"],
            font=dict(family="Inter, Segoe UI, sans-serif", color=C["text"], size=12),
            margin=dict(l=55, r=30, t=60, b=45))

def style(fig):
    fig.update_xaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    fig.update_yaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    return fig

# ── Generate data ──
X_dash, y_dash = make_blobs(n_samples=300, centers=4, cluster_std=0.9, random_state=42)

figs = {}

# ── TAB 1: K-Means ──

# Animated cluster assignment: show K-Means iterations as frames
def run_kmeans_steps(X, K, max_iter=8):
    np.random.seed(42)
    idx = np.random.choice(len(X), K, replace=False)
    centroids = X[idx].copy()
    frames_data = []
    for it in range(max_iter):
        dists = np.array([np.sum((X - c)**2, axis=1) for c in centroids]).T
        labels = np.argmin(dists, axis=1)
        frames_data.append({'labels': labels.copy(), 'centroids': centroids.copy(), 'iter': it})
        new_c = np.array([X[labels==k].mean(axis=0) if np.sum(labels==k)>0 else centroids[k] for k in range(K)])
        if np.allclose(centroids, new_c):
            break
        centroids = new_c
    return frames_data

steps = run_kmeans_steps(X_dash, 4)

# Build animated figure with slider
f_km = go.Figure()

# Initial frame data
init_step = steps[0]
for k in range(4):
    mask = init_step['labels'] == k
    f_km.add_trace(go.Scatter(
        x=X_dash[mask, 0], y=X_dash[mask, 1], mode='markers',
        marker=dict(size=6, color=C['palette'][k], opacity=0.7),
        name=f'Cluster {k+1}', showlegend=True))
f_km.add_trace(go.Scatter(
    x=init_step['centroids'][:, 0], y=init_step['centroids'][:, 1],
    mode='markers', marker=dict(size=16, color='white', symbol='x',
    line=dict(width=2, color='black')), name='Centroids'))

# Create frames
frames = []
for step in steps:
    frame_traces = []
    for k in range(4):
        mask = step['labels'] == k
        frame_traces.append(go.Scatter(
            x=X_dash[mask, 0], y=X_dash[mask, 1], mode='markers',
            marker=dict(size=6, color=C['palette'][k], opacity=0.7)))
    frame_traces.append(go.Scatter(
        x=step['centroids'][:, 0], y=step['centroids'][:, 1],
        mode='markers', marker=dict(size=16, color='white', symbol='x',
        line=dict(width=2, color='black'))))
    frames.append(go.Frame(data=frame_traces, name=f'Iter {step["iter"]}'))

f_km.frames = frames
sliders = [dict(active=0, steps=[dict(args=[[f'Iter {s["iter"]}'],
    dict(frame=dict(duration=500, redraw=True), mode='immediate')],
    label=f'{s["iter"]}', method='animate') for s in steps],
    x=0.1, len=0.8, y=0, currentvalue=dict(prefix='Iteration: ',
    font=dict(size=14, color=C['text'])),
    font=dict(color=C['muted']),
    bgcolor=C['grid'], activebgcolor=C['accent'])]

f_km.update_layout(**BASE, height=480,
    title=dict(text='<b>K-Means Iterations</b> (K=4)', font=dict(size=16)),
    sliders=sliders,
    updatemenus=[dict(type='buttons', showactive=False, x=0.05, y=1.12,
        buttons=[dict(label='Play', method='animate',
            args=[None, dict(frame=dict(duration=700, redraw=True),
                  fromcurrent=True, mode='immediate')]),
            dict(label='Pause', method='animate',
            args=[[None], dict(frame=dict(duration=0, redraw=False),
                  mode='immediate')])])])
style(f_km)
figs['km_anim'] = f_km

# Elbow plot
inertias_dash = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X_dash)
    inertias_dash.append(km.inertia_)

f_elbow = go.Figure()
f_elbow.add_trace(go.Scatter(x=list(range(1,11)), y=inertias_dash, mode='lines+markers',
    marker=dict(size=9, color=C['accent']), line=dict(color=C['accent'], width=3)))
f_elbow.add_vline(x=4, line_dash='dash', line_color=C['green'],
    annotation_text='Elbow at K=4', annotation_font_color=C['green'])
f_elbow.update_layout(**BASE, height=380,
    title=dict(text='<b>Elbow Method</b>', font=dict(size=16)),
    xaxis_title='K', yaxis_title='Inertia (WCSS)')
style(f_elbow)
figs['elbow'] = f_elbow

# ── TAB 2: Hierarchical ──

# Dendrogram as plotly figure (using scipy dendrogram output)
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

X_hc_dash = X_dash[:60]  # Use subset for readable dendrogram

def make_plotly_dendrogram(X, method, color):
    Z = linkage(X, method=method)
    fig_tmp, ax_tmp = plt.subplots()
    dn = dendrogram(Z, ax=ax_tmp, no_plot=False)
    plt.close(fig_tmp)
    
    f = go.Figure()
    for xs, ys in zip(dn['icoord'], dn['dcoord']):
        f.add_trace(go.Scatter(x=xs, y=ys, mode='lines',
            line=dict(color=color, width=1.5), showlegend=False))
    f.update_layout(**BASE, height=400,
        title=dict(text=f'<b>{method.capitalize()} Linkage Dendrogram</b>', font=dict(size=16)),
        xaxis_title='Sample Index', yaxis_title='Distance')
    style(f)
    return f

figs['dend_ward'] = make_plotly_dendrogram(X_hc_dash, 'ward', C['accent'])

# Linkage comparison: scatter plots
f_link = make_subplots(rows=2, cols=2,
    subplot_titles=['Single', 'Complete', 'Average', 'Ward'],
    horizontal_spacing=0.08, vertical_spacing=0.12)

for idx, method in enumerate(['single', 'complete', 'average', 'ward']):
    row = idx // 2 + 1
    col = idx % 2 + 1
    Z_tmp = linkage(X_dash, method=method)
    lab_tmp = fcluster(Z_tmp, t=4, criterion='maxclust') - 1
    for k in range(4):
        mask = lab_tmp == k
        f_link.add_trace(go.Scatter(
            x=X_dash[mask, 0], y=X_dash[mask, 1], mode='markers',
            marker=dict(size=4, color=C['palette'][k], opacity=0.7),
            showlegend=False), row, col)

f_link.update_layout(**BASE, height=600,
    title=dict(text='<b>Linkage Comparison</b> (K=4)', font=dict(size=16)))
style(f_link)
figs['linkage'] = f_link

# ── TAB 3: Validation (Silhouette) ──

km_final = KMeans(n_clusters=4, n_init=10, random_state=42).fit(X_dash)
sil_vals = silhouette_samples(X_dash, km_final.labels_)
sil_avg = silhouette_score(X_dash, km_final.labels_)

f_sil = go.Figure()
y_lower = 0
for k in range(4):
    cluster_sil = np.sort(sil_vals[km_final.labels_ == k])
    cluster_size = len(cluster_sil)
    y_upper = y_lower + cluster_size
    f_sil.add_trace(go.Bar(
        x=cluster_sil, y=list(range(y_lower, y_upper)),
        orientation='h', marker_color=C['palette'][k],
        name=f'Cluster {k+1}', showlegend=True))
    y_lower = y_upper + 5  # gap between clusters

f_sil.add_vline(x=sil_avg, line_dash='dash', line_color=C['pink'],
    annotation_text=f'Avg: {sil_avg:.3f}', annotation_font_color=C['pink'])
f_sil.update_layout(**BASE, height=500,
    title=dict(text='<b>Silhouette Plot</b> (K=4)', font=dict(size=16)),
    xaxis_title='Silhouette Coefficient', yaxis_title='Samples (sorted)',
    yaxis=dict(showticklabels=False), barmode='stack', bargap=0)
style(f_sil)
figs['silhouette'] = f_sil

# ── KPI indicators ──
f_kpi = make_subplots(rows=1, cols=4, specs=[[{'type':'indicator'}]*4], horizontal_spacing=0.05)
kpi_data = [
    ('Inertia', km_final.inertia_, '.1f'),
    ('Silhouette', sil_avg, '.3f'),
    ('K', 4, 'd'),
    ('n', len(X_dash), 'd')
]
for i, (name, val, fmt) in enumerate(kpi_data, 1):
    f_kpi.add_trace(go.Indicator(mode='number', value=val,
        number=dict(valueformat=fmt, font=dict(size=36, color=C['text'])),
        title=dict(text=f"<span style='color:{C['muted']};font-size:14px'>{name}</span>")), 1, i)
f_kpi.update_layout(**BASE, height=160)
figs['kpi'] = f_kpi

# ── Assemble dashboard ──
def div(key, first=False):
    return figs[key].to_html(full_html=False,
        include_plotlyjs='cdn' if first else False,
        config={'displayModeBar': True, 'displaylogo': False})

card = lambda inner: f"<div class='card'>{inner}</div>"
tabs = {
    'K-Means': card(div('kpi', first=True)) + card(div('km_anim')) + card(div('elbow')),
    'Hierarchical': card(div('dend_ward')) + card(div('linkage')),
    'Validation': card(div('silhouette')),
}
nav = ''.join(f"<button class='tab-btn{' active' if i==0 else ''}' onclick=\"showTab(event,'tab{i}')\">"+name+'</button>' for i, name in enumerate(tabs))
panels = ''.join(f"<div id='tab{i}' class='tab-panel' style='display:{'block' if i==0 else 'none'}'>{c}</div>" for i, c in enumerate(tabs.values()))

html = f"""<div id="dash"><style>
  #dash {{background:{C['bg']};border-radius:14px;padding:22px;font-family:Inter,'Segoe UI',sans-serif;}}
  #dash .dash-header {{display:flex;justify-content:space-between;align-items:baseline;margin-bottom:16px;}}
  #dash h1 {{color:{C['text']};font-size:21px;margin:0;}}
  #dash .sub {{color:{C['muted']};font-size:12px;}}
  #dash .tab-bar {{display:flex;gap:6px;border-bottom:1px solid {C['grid']};margin-bottom:18px;}}
  #dash .tab-btn {{background:none;border:none;color:{C['muted']};padding:10px 18px;font-size:13.5px;font-weight:600;cursor:pointer;border-bottom:2.5px solid transparent;}}
  #dash .tab-btn:hover {{color:{C['text']};}}
  #dash .tab-btn.active {{color:{C['accent']};border-bottom-color:{C['accent']};}}
  #dash .card {{background:{C['card']};border:1px solid {C['grid']};border-radius:12px;padding:6px;margin-bottom:16px;box-shadow:0 2px 10px rgba(0,0,0,.35);}}
</style>
<div class="dash-header"><h1>Clustering Explorer</h1><span class="sub">BDATA 412 \u00b7 Section 3</span></div>
<div class="tab-bar">{nav}</div>{panels}
<script>function showTab(evt,id){{const d=document.getElementById('dash');d.querySelectorAll('.tab-panel').forEach(p=>p.style.display='none');d.querySelectorAll('.tab-btn').forEach(b=>b.classList.remove('active'));document.getElementById(id).style.display='block';evt.currentTarget.classList.add('active');document.getElementById(id).querySelectorAll('.js-plotly-plot').forEach(g=>Plotly.Plots.resize(g));}}</script></div>"""
display(HTML(html))
print('Dashboard ready \u2014 switch tabs to explore K-Means, Hierarchical Clustering, and Validation.')
```
