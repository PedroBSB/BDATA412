<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Matrix-Vector Multiplication</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Each output entry is the <strong>dot product</strong> of one row of A with x. Dimensions: (m×n)·(n×1) = (m×1). This is the heart of neural networks: y = Wx + b.</p>
</div>

```python
# Matrix-vector multiplication
A = np.array([[1, 2, 3],
              [4, 5, 6]])
x = np.array([1, 0, -1])
y = A @ x
print(f"A (2×3) · x (3×1) = y (2×1)")
print(f"Row 1 · x: (1)(1) + (2)(0) + (3)(-1) = {y[0]}")
print(f"Row 2 · x: (4)(1) + (5)(0) + (6)(-1) = {y[1]}")
print(f"y = {y}")
```

```python
# Visualize geometric transformations
theta_vals = np.linspace(0, 2*np.pi, 100)
circle = np.array([np.cos(theta_vals), np.sin(theta_vals)])

transforms = {
    'Rotate 90°': np.array([[0,-1],[1,0]]),
    'Scale 1.5×': np.array([[1.5,0],[0,1.5]]),
    'Shear': np.array([[1,1.5],[0,1]]),
    'Project to x': np.array([[1,0],[0,0]]),
}
fig = make_subplots(rows=1, cols=4, subplot_titles=list(transforms.keys()))
for i, (name, M) in enumerate(transforms.items(), 1):
    transformed = M @ circle
    fig.add_trace(go.Scatter(x=circle[0], y=circle[1], mode='lines',
        line=dict(color='#8B90A0', width=1), showlegend=False), 1, i)
    fig.add_trace(go.Scatter(x=transformed[0], y=transformed[1], mode='lines',
        line=dict(color='#4B2E83', width=2.5), showlegend=False), 1, i)
fig.update_layout(height=300, template='plotly_white', title='<b>Matrices as Geometric Actions</b>',
                 font=dict(family='Encode Sans, Arial, sans-serif'))
for i in range(1, 5):
    fig.update_xaxes(scaleanchor=f'y{i if i>1 else ""}', range=[-2.5,2.5], row=1, col=i)
    fig.update_yaxes(range=[-2.5,2.5], row=1, col=i)
fig.show()
```

```python
# Determinant examples
A = np.array([[3, 2], [1, 4]])
print(f"A = {A.tolist()}")
print(f"det(A) = (3)(4) - (2)(1) = {np.linalg.det(A):.0f}")
print(f"|det(A)| = {abs(np.linalg.det(A)):.0f} → A scales area by {abs(np.linalg.det(A)):.0f}×")

B = np.array([[2, 6], [1, 3]])
print(f"\nB = {B.tolist()}")
print(f"det(B) = (2)(3) - (6)(1) = {np.linalg.det(B):.0f}")
print(f"B is {'singular (not invertible)' if np.isclose(np.linalg.det(B), 0) else 'invertible'}")
```

```python
# Solving Ax = b
A = np.array([[1, 2], [3, 5]])
b = np.array([4, 13])
x = np.linalg.solve(A, b)
print(f"Solve Ax = b where A = {A.tolist()}, b = {b}")
print(f"x = {x}")
print(f"Verify: A·x = {A @ x}")
```

<div style="background: linear-gradient(135deg, #4B2E83 0%, #32006E 100%); padding: 30px; border-radius: 10px; margin: 40px 0 30px 0; box-shadow: 0 5px 15px rgba(75,46,131,0.2);">
  <h1 style="color: #FFF; font-family: 'Encode Sans', 'Arial', sans-serif; font-weight: 700; margin: 0; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);">Part 3: Covariance &amp; Correlation Matrices</h1>
  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Measuring relationships between features</p>
</div>

```python
# Worked example: 5×2 data
X = np.array([[2,8],[4,6],[6,9],[8,5],[10,7]])
mu = X.mean(axis=0)
Xc = X - mu
S = (Xc.T @ Xc) / (len(X) - 1)
sigma = np.sqrt(np.diag(S))
R = S / np.outer(sigma, sigma)

print("Data X:")
print(pd.DataFrame(X, columns=['x1','x2']))
print(f"\nMean: μ = {mu}")
print(f"\nCovariance S:\n{pd.DataFrame(S, columns=['x1','x2'], index=['x1','x2']).round(2)}")
print(f"\nCorrelation R:\n{pd.DataFrame(R, columns=['x1','x2'], index=['x1','x2']).round(2)}")
```

```python
# Correlation heatmap
fig = go.Figure(go.Heatmap(
    z=R, x=['x1','x2'], y=['x1','x2'],
    colorscale=[[0,'#C62828'],[0.5,'#FFFFFF'],[1,'#4B2E83']],
    zmin=-1, zmax=1,
    text=np.round(R, 2), texttemplate='%{text}', textfont=dict(size=16)))
fig.update_layout(title='<b>Correlation Matrix Heatmap</b>',
    template='plotly_white', height=350, width=400,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Why Standardize?</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Without scaling, the biggest-magnitude feature wins. Ann (age=25, income=$40,000) vs Bob (age=45, income=$42,000): raw distance ≈ 2000 (99.99% income). After z-scores: distance ≈ 2.0 (99.6% age). <strong>z = (x − μ) / σ</strong> puts every feature on the same scale.</p>
</div>

```python
# Standardization example
raw = np.array([[25, 40000], [45, 42000], [30, 55000], [50, 38000]])
labels = ['Ann','Bob','Carol','Dave']
mu = raw.mean(axis=0)
sigma = raw.std(axis=0, ddof=1)
Z = (raw - mu) / sigma
print("Raw data:")
print(pd.DataFrame(raw, index=labels, columns=['Age','Income']))
print(f"\nStandardized (z-scores):")
print(pd.DataFrame(Z.round(3), index=labels, columns=['Age','Income']))
print(f"\nRaw distance Ann-Bob: {np.linalg.norm(raw[0]-raw[1]):.0f}")
print(f"Standardized distance Ann-Bob: {np.linalg.norm(Z[0]-Z[1]):.2f}")
```
