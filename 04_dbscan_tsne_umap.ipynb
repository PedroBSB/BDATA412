  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Density-based clustering that finds arbitrarily shaped clusters</p>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Core, Border &amp; Noise Points</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">DBSCAN (Density-Based Spatial Clustering of Applications with Noise) classifies every point into one of three categories:</p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.7;"><li><strong>Core point:</strong> has at least <em>minPts</em> neighbors within radius ε (epsilon). These form the dense backbone of a cluster.</li><li><strong>Border point:</strong> falls within ε of a core point but has fewer than minPts neighbors itself. It belongs to the cluster but sits on its edge.</li><li><strong>Noise point:</strong> neither core nor border — isolated outliers that belong to no cluster (labeled -1).</li></ul>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Key advantage:</strong> DBSCAN does not require specifying the number of clusters in advance, and it naturally handles outliers.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Epsilon (ε) and MinPts Parameters</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>ε (eps):</strong> the radius of the neighborhood around each point. Too small → everything is noise; too large → everything merges into one cluster.</p>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>minPts:</strong> minimum number of points required within ε to qualify as a core point. Rule of thumb: minPts ≥ dimensionality + 1. For 2D data, minPts = 5 is a common starting point.</p>
</div>

```python
# DBSCAN on moons dataset — where k-means fails
np.random.seed(42)
X_moons, y_moons = make_moons(n_samples=500, noise=0.08, random_state=42)

# k-means (fails on non-convex shapes)
km = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X_moons)

# DBSCAN (handles non-convex shapes)
db = DBSCAN(eps=0.2, min_samples=5).fit(X_moons)

print(f"k-Means clusters: {len(set(km.labels_))}")
print(f"DBSCAN clusters:  {len(set(db.labels_) - {-1})}")
print(f"DBSCAN noise pts: {(db.labels_ == -1).sum()}")

# Side-by-side comparison
fig = make_subplots(rows=1, cols=2, subplot_titles=(
    '<b>k-Means (fails)</b>', '<b>DBSCAN (succeeds)</b>'))

colors_km = ['#4B2E83' if l == 0 else '#E8B923' for l in km.labels_]
fig.add_trace(go.Scatter(x=X_moons[:, 0], y=X_moons[:, 1], mode='markers',
    marker=dict(size=5, color=colors_km), showlegend=False), 1, 1)

color_map = {-1: '#999999', 0: '#4B2E83', 1: '#E8B923'}
colors_db = [color_map.get(l, '#22D3A5') for l in db.labels_]
fig.add_trace(go.Scatter(x=X_moons[:, 0], y=X_moons[:, 1], mode='markers',
    marker=dict(size=5, color=colors_db), showlegend=False), 1, 2)

fig.update_layout(template='plotly_white', height=400,
    title='<b>Non-Convex Clusters:</b> k-Means vs DBSCAN on Moons Data',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

```python
# DBSCAN on concentric circles — another non-convex example
np.random.seed(42)
X_circles, y_circles = make_circles(n_samples=500, noise=0.05, factor=0.5, random_state=42)

db_circles = DBSCAN(eps=0.2, min_samples=5).fit(X_circles)
km_circles = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X_circles)

fig = make_subplots(rows=1, cols=2, subplot_titles=(
    '<b>k-Means (fails)</b>', '<b>DBSCAN (succeeds)</b>'))

colors_km2 = ['#4B2E83' if l == 0 else '#E8B923' for l in km_circles.labels_]
fig.add_trace(go.Scatter(x=X_circles[:, 0], y=X_circles[:, 1], mode='markers',
    marker=dict(size=5, color=colors_km2), showlegend=False), 1, 1)

color_map2 = {-1: '#999999', 0: '#4B2E83', 1: '#E8B923'}
colors_db2 = [color_map2.get(l, '#22D3A5') for l in db_circles.labels_]
fig.add_trace(go.Scatter(x=X_circles[:, 0], y=X_circles[:, 1], mode='markers',
    marker=dict(size=5, color=colors_db2), showlegend=False), 1, 2)

fig.update_layout(template='plotly_white', height=400,
    title='<b>Concentric Circles:</b> k-Means vs DBSCAN',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">k-Distance Plot for Choosing ε</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The <strong>k-distance plot</strong> helps select ε empirically. For each point, compute the distance to its k-th nearest neighbor (where k = minPts). Sort these distances and plot them. The “elbow” in the curve suggests a good ε: points below the elbow are in dense regions (clusters), points above are noise.</p>
</div>

```python
# k-Distance plot for epsilon selection
k = 5  # minPts
nn = NearestNeighbors(n_neighbors=k)
nn.fit(X_moons)
distances, _ = nn.kneighbors(X_moons)

# Sort k-th neighbor distances (last column)
k_dists = np.sort(distances[:, -1])[::-1]

fig = go.Figure()
fig.add_trace(go.Scatter(y=k_dists, mode='lines',
    line=dict(color='#4B2E83', width=2.5), name='k-distance'))

# Mark the elbow region
fig.add_hline(y=0.2, line_dash='dash', line_color='#C62828',
    annotation_text='ε = 0.2 (elbow)', annotation_position='top right')

fig.update_layout(
    title=f'<b>k-Distance Plot</b> (k={k}) — elbow indicates optimal ε',
    xaxis_title='Points (sorted by distance)',
    yaxis_title=f'{k}-th Nearest Neighbor Distance',
    template='plotly_white', height=400,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
print(f"The elbow around ε ≈ 0.2 separates dense cluster regions from sparse noise.")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Non-Convex Shape Advantage</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">k-Means assigns points to the nearest centroid, producing <strong>convex (Voronoi) partitions</strong>. This fails for crescent moons, rings, or any non-globular shape. DBSCAN grows clusters by chaining core points through density connectivity, so it can trace <strong>any shape</strong> as long as it is dense enough. No assumption about cluster geometry is needed.</p>
</div>

```python
# Demonstrate effect of varying epsilon
np.random.seed(42)
eps_values = [0.1, 0.2, 0.3, 0.5]
fig = make_subplots(rows=1, cols=4, subplot_titles=[f'<b>ε = {e}</b>' for e in eps_values])

for i, eps in enumerate(eps_values, 1):
    db_i = DBSCAN(eps=eps, min_samples=5).fit(X_moons)
    n_clusters = len(set(db_i.labels_) - {-1})
    n_noise = (db_i.labels_ == -1).sum()
    cmap = {-1: '#999999', 0: '#4B2E83', 1: '#E8B923', 2: '#22D3A5', 3: '#E8608A'}
    cols = [cmap.get(l, '#4F8EF7') for l in db_i.labels_]
    fig.add_trace(go.Scatter(x=X_moons[:, 0], y=X_moons[:, 1], mode='markers',
        marker=dict(size=4, color=cols), showlegend=False,
        text=[f'Cluster: {l}' for l in db_i.labels_]), 1, i)

fig.update_layout(template='plotly_white', height=350,
    title='<b>Effect of ε on DBSCAN Clustering</b> (minPts=5)',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">High-D vs Low-D Affinities</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">t-SNE (t-distributed Stochastic Neighbor Embedding) works in two stages:</p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.7;"><li><strong>High-D:</strong> Convert pairwise distances to conditional probabilities p<sub>j|i</sub> using Gaussian kernels. Nearby points get high probability; distant points get near-zero probability.</li><li><strong>Low-D:</strong> Place points in 2D and compute similarities q<sub>ij</sub> using a <em>Student-t distribution</em> (heavy tails). The heavier tails help prevent the “crowding problem” by allowing moderately distant points to be placed farther apart in the map.</li></ul>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Perplexity</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Perplexity controls the effective number of neighbors each point considers. It is related to the bandwidth of the Gaussian kernel: <strong>Perp = 2<sup>H(P<sub>i</sub>)</sup></strong> where H is the entropy. Low perplexity (5–10) focuses on very local structure; high perplexity (30–50) captures more global patterns. Typical range: 5–50. Results can change significantly with perplexity — always try multiple values.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">KL Divergence Cost &amp; Gradient</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">t-SNE minimizes the <strong>Kullback-Leibler divergence</strong> between the high-D distribution P and the low-D distribution Q:</p>
  <p style="font-size: 1.1em; color: #333; text-align: center;"><strong>C = KL(P || Q) = Σ<sub>i</sub> Σ<sub>j</sub> p<sub>ij</sub> log(p<sub>ij</sub> / q<sub>ij</sub>)</strong></p>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">KL divergence is <strong>asymmetric</strong>: it penalizes placing nearby high-D points far apart in low-D (large p, small q) much more than placing distant high-D points close in low-D. This means t-SNE <strong>preserves local neighborhoods</strong> well but may distort global distances and cluster sizes.</p>
</div>

```python
# t-SNE on Iris with perplexity comparison
iris = load_iris()
X_iris = iris.data
y_iris = iris.target
target_names = iris.target_names

perplexities = [5, 15, 30, 50]
fig = make_subplots(rows=1, cols=4, subplot_titles=[f'<b>Perplexity = {p}</b>' for p in perplexities])

colors_iris = {0: '#4B2E83', 1: '#E8B923', 2: '#22D3A5'}

for i, perp in enumerate(perplexities, 1):
    tsne = TSNE(n_components=2, perplexity=perp, random_state=42, n_iter=1000)
    X_tsne = tsne.fit_transform(X_iris)
    for cls in range(3):
        mask = y_iris == cls
        fig.add_trace(go.Scatter(
            x=X_tsne[mask, 0], y=X_tsne[mask, 1], mode='markers',
            marker=dict(size=5, color=colors_iris[cls]),
            name=target_names[cls], showlegend=(i == 1),
            legendgroup=target_names[cls]), 1, i)

fig.update_layout(template='plotly_white', height=400,
    title='<b>t-SNE on Iris</b> — Effect of Perplexity',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
print("Low perplexity → tight local clusters. High perplexity → broader structure emerges.")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Why t-SNE is Good for Visualization but Not Distances</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">t-SNE excels at revealing <strong>clusters and local neighborhoods</strong> in a 2D scatter plot. However, there are important caveats:</p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.7;"><li><strong>Distances between clusters</strong> are not meaningful — two well-separated groups may appear at arbitrary distances.</li><li><strong>Cluster sizes</strong> do not reflect the true density or spread of the data.</li><li><strong>Non-deterministic:</strong> different random seeds produce different layouts.</li><li><strong>Slow</strong> for large datasets (O(n²) in the naive implementation).</li></ul>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>Bottom line:</strong> use t-SNE for <em>exploration and visualization</em>, not for measuring distances or feeding into downstream models.</p>
</div>

  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Uniform Manifold Approximation and Projection</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Local Connectivity ρ</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">UMAP assumes the data is uniformly distributed on a <strong>Riemannian manifold</strong>. For each point, ρ<sub>i</sub> is the distance to its nearest neighbor. UMAP subtracts ρ<sub>i</sub> from all distances, ensuring every point is connected to at least one neighbor with edge weight 1. This <strong>local connectivity</strong> assumption prevents isolated points and adapts the distance scale per point.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Fuzzy Simplicial Set Union</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Each point builds a <strong>local fuzzy graph</strong> based on its k nearest neighbors (weighted by exponentially decaying distance). These local graphs are merged via a <strong>fuzzy union</strong>: w<sub>ij</sub> = w<sub>i→j</sub> + w<sub>j→i</sub> - w<sub>i→j</sub> · w<sub>j→i</sub>. This symmetrization creates the global topological representation (a fuzzy simplicial set).</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Low-D Layout &amp; Cross-Entropy Cost</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">UMAP optimizes a <strong>fuzzy set cross-entropy</strong> between the high-D graph and its low-D counterpart:</p>
  <p style="font-size: 1.1em; color: #333; text-align: center;"><strong>C = Σ<sub>ij</sub> [ w<sub>ij</sub> log(w<sub>ij</sub>/v<sub>ij</sub>) + (1-w<sub>ij</sub>) log((1-w<sub>ij</sub>)/(1-v<sub>ij</sub>)) ]</strong></p>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Unlike KL divergence in t-SNE, cross-entropy has <strong>both attractive and repulsive terms</strong>. The first term attracts connected points together; the second term pushes unconnected points apart. This gives UMAP better preservation of <strong>global structure</strong> compared to t-SNE.</p>
</div>

```python
# UMAP on Iris
X_scaled_iris = StandardScaler().fit_transform(X_iris)

reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=42)
X_umap = reducer.fit_transform(X_scaled_iris)

fig = go.Figure()
for cls in range(3):
    mask = y_iris == cls
    fig.add_trace(go.Scatter(
        x=X_umap[mask, 0], y=X_umap[mask, 1], mode='markers',
        marker=dict(size=7, color=colors_iris[cls], opacity=0.8),
        name=target_names[cls]))

fig.update_layout(template='plotly_white', height=450,
    title='<b>UMAP on Iris</b> (n_neighbors=15, min_dist=0.1)',
    xaxis_title='UMAP-1', yaxis_title='UMAP-2',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Comparison with t-SNE</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Key differences between UMAP and t-SNE:</p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.7;"><li><strong>Speed:</strong> UMAP is significantly faster (often 10–100x for large datasets).</li><li><strong>Global structure:</strong> UMAP better preserves relative positions of clusters.</li><li><strong>Determinism:</strong> UMAP with a fixed seed is more reproducible.</li><li><strong>Cost function:</strong> UMAP uses cross-entropy (symmetric); t-SNE uses KL divergence (asymmetric).</li><li><strong>Scalability:</strong> UMAP handles millions of points; t-SNE struggles beyond ~10k.</li></ul>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Both are <strong>non-linear</strong> and excellent for visualization. UMAP has become the default choice for many practitioners due to its speed and global coherence.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Silhouette Score</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The silhouette score measures how similar a point is to its own cluster vs. the nearest neighboring cluster:</p>
  <p style="font-size: 1.1em; color: #333; text-align: center;"><strong>s(i) = (b(i) - a(i)) / max(a(i), b(i))</strong></p>
  <ul style="font-size: 1.05em; color: #555; line-height: 1.7;"><li><strong>a(i):</strong> mean distance from point i to all other points in the same cluster (cohesion).</li><li><strong>b(i):</strong> mean distance from point i to all points in the nearest different cluster (separation).</li><li>Range: [-1, 1]. Near +1 = well-clustered. Near 0 = on boundary. Near -1 = misassigned.</li></ul>
</div>

```python
# Silhouette analysis on Iris with KMeans
from sklearn.metrics import silhouette_score, silhouette_samples

X_sil = StandardScaler().fit_transform(X_iris)

# Try different k values
k_range = range(2, 7)
scores = []
for k in k_range:
    km_k = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_sil)
    s = silhouette_score(X_sil, km_k.labels_)
    scores.append(s)
    print(f"k={k}: silhouette = {s:.4f}")

fig = go.Figure(go.Bar(x=list(k_range), y=scores,
    marker_color=['#E8B923' if s == max(scores) else '#4B2E83' for s in scores],
    text=[f'{s:.3f}' for s in scores], textposition='outside'))
fig.update_layout(template='plotly_white', height=400,
    title='<b>Silhouette Score</b> vs Number of Clusters (k-Means on Iris)',
    xaxis_title='k (number of clusters)', yaxis_title='Silhouette Score',
    font=dict(family='Encode Sans, Arial, sans-serif'),
    yaxis=dict(range=[0, max(scores) + 0.1]))
fig.show()
print(f"\nBest k = {list(k_range)[np.argmax(scores)]} (highest silhouette)")
```

```python
# Silhouette plot per sample (k=3)
km3 = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_sil)
sample_sil = silhouette_samples(X_sil, km3.labels_)
avg_sil = silhouette_score(X_sil, km3.labels_)

fig = go.Figure()
y_lower = 0
cluster_colors = ['#4B2E83', '#E8B923', '#22D3A5']

for cls in range(3):
    cls_sil = np.sort(sample_sil[km3.labels_ == cls])
    size_cls = len(cls_sil)
    y_upper = y_lower + size_cls
    fig.add_trace(go.Bar(
        x=cls_sil, y=list(range(y_lower, y_upper)),
        orientation='h', name=f'Cluster {cls}',
        marker_color=cluster_colors[cls], width=1.0))
    y_lower = y_upper + 5

fig.add_vline(x=avg_sil, line_dash='dash', line_color='#C62828',
    annotation_text=f'Avg = {avg_sil:.3f}')
fig.update_layout(template='plotly_white', height=500,
    title='<b>Silhouette Plot</b> (k=3, k-Means on Iris)',
    xaxis_title='Silhouette Coefficient', yaxis_title='Sample Index',
    font=dict(family='Encode Sans, Arial, sans-serif'),
    showlegend=True, barmode='stack')
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">F-Statistic (ANOVA)</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">The F-statistic from one-way ANOVA tests whether the cluster means are significantly different for each feature:</p>
  <p style="font-size: 1.1em; color: #333; text-align: center;"><strong>F = (Between-group variance) / (Within-group variance)</strong></p>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">A large F means the feature strongly separates the clusters. This helps identify which features are most important for distinguishing your clusters.</p>
</div>

```python
# F-statistic for each feature
from scipy.stats import f_oneway

feature_names = iris.feature_names
f_stats = []
for j, fname in enumerate(feature_names):
    groups = [X_iris[km3.labels_ == c, j] for c in range(3)]
    f_val, p_val = f_oneway(*groups)
    f_stats.append({'Feature': fname, 'F-statistic': round(f_val, 2), 'p-value': f'{p_val:.2e}'})
    print(f"{fname:25s}  F = {f_val:8.2f}  p = {p_val:.2e}")

df_f = pd.DataFrame(f_stats)
fig = go.Figure(go.Bar(
    x=df_f['Feature'], y=df_f['F-statistic'],
    marker_color='#4B2E83',
    text=df_f['F-statistic'], textposition='outside'))
fig.update_layout(template='plotly_white', height=400,
    title='<b>F-Statistic per Feature</b> — Which features separate clusters?',
    xaxis_title='Feature', yaxis_title='F-statistic',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Centroid z-Profiles &amp; Naming Clusters as Personas</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">To interpret clusters, compute the <strong>centroid</strong> of each cluster, then standardize across features to get <strong>z-score profiles</strong>. Each cluster’s z-profile reveals which features are above or below average. These profiles can be given <strong>persona names</strong> that summarize their characteristics, making results actionable for stakeholders.</p>
</div>

```python
# Centroid z-score profiles
df_iris = pd.DataFrame(X_iris, columns=feature_names)
df_iris['Cluster'] = km3.labels_

# Compute cluster centroids
centroids = df_iris.groupby('Cluster')[feature_names].mean()

# z-score the centroids (standardize across clusters for each feature)
z_centroids = (centroids - centroids.mean()) / centroids.std()
print("Cluster centroids (raw):")
print(centroids.round(2))
print("\nCluster centroids (z-scored):")
print(z_centroids.round(2))

# Bar chart of z-profiles
fig = go.Figure()
persona_names = ['Small & Compact', 'Large & Wide', 'Medium & Narrow']
cluster_colors = ['#4B2E83', '#E8B923', '#22D3A5']

for cls in range(3):
    fig.add_trace(go.Bar(
        x=feature_names, y=z_centroids.loc[cls].values,
        name=f'Cluster {cls}: {persona_names[cls]}',
        marker_color=cluster_colors[cls]))

fig.update_layout(template='plotly_white', height=450, barmode='group',
    title='<b>Cluster Profiles</b> — Centroid z-Scores by Feature',
    xaxis_title='Feature', yaxis_title='z-Score (relative to cluster means)',
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.add_hline(y=0, line_color='gray', line_dash='dash')
fig.show()
print("\nPersona labels help stakeholders understand what each cluster represents.")
```

  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Run DBSCAN on the moons dataset (eps=0.2, min_samples=5). Compute the silhouette score, excluding noise points (label=-1). Compare with the k-Means (k=2) silhouette score on the same data. Which method produces better-defined clusters?</p>
