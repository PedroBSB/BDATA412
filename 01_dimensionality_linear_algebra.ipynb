<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Distance Concentration</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">In high dimensions, all pairwise distances converge — max ≈ min. Distance-based methods (k-NN, k-means, clustering) lose discriminating power. <strong>Rule of thumb:</strong> if nearest and farthest neighbors are within 10% of each other, distance is no longer a useful signal.</p>
</div>

```python
# Demonstrate distance concentration across dimensions
from scipy.spatial.distance import pdist

dims = [1, 2, 5, 10, 50, 100, 500, 1000]
n_points = 100
ratios = []

for d in dims:
    X = np.random.uniform(0, 1, (n_points, d))
    dists = pdist(X, 'euclidean')
    ratio = dists.max() / dists.min()
    ratios.append(ratio)
    print(f"d={d:>5d}:  max/min ratio = {ratio:.2f}")

fig = go.Figure(go.Scatter(
    x=dims, y=ratios, mode='lines+markers',
    marker=dict(size=10, color='#4B2E83'),
    line=dict(color='#4B2E83', width=3)))
fig.update_layout(
    title='<b>Distance Concentration</b> — max/min ratio collapses in high-D',
    xaxis_title='Number of Dimensions',
    yaxis_title='max(dist) / min(dist)',
    template='plotly_white', height=400,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.add_hline(y=1, line_dash="dash", line_color="#C62828",
             annotation_text="All equidistant (ratio = 1)")
fig.show()
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Sparsity</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Data needed to cover the sample space grows <strong>exponentially</strong> with dimensions. 10 points in 1D feel dense; the same 10 points in 100D are isolated in a vast empty space. <strong>Fix:</strong> feature selection or feature extraction (PCA, LDA).</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Nearest-Neighbor Breakdown</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">When all points are equidistant, 'nearest' is just noise. k-NN, k-means, and clustering all rely on a meaningful gap between nearest and farthest neighbor. In high-D, that gap vanishes.</p>
</div>

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Overfitting Risk</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">More features = more model complexity. The model starts learning noise instead of signal. <strong>Fix:</strong> regularization (Lasso L₁, Ridge L₂), tree-based models, or reduce dimensions first.</p>
</div>

<div style="background: linear-gradient(135deg, #4B2E83 0%, #32006E 100%); padding: 30px; border-radius: 10px; margin: 40px 0 30px 0; box-shadow: 0 5px 15px rgba(75,46,131,0.2);">
  <h1 style="color: #FFF; font-family: 'Encode Sans', 'Arial', sans-serif; font-weight: 700; margin: 0; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);">Part 2: K-Nearest Neighbors (KNN)</h1>
  <p style="color: #E8D3FF; font-family: 'Encode Sans', 'Arial', sans-serif; font-size: 1.3em; margin: 10px 0; font-weight: 300;">Classification by majority vote of K nearest neighbors</p>
</div>

```python
# KNN from scratch
def knn_classify(X_train, y_train, query, k=3):
    """Classify query point using k-NN majority vote."""
    distances = np.sqrt(np.sum((X_train - query)**2, axis=1))
    nearest_idx = np.argsort(distances)[:k]
    nearest_labels = y_train[nearest_idx]
    nearest_dists = distances[nearest_idx]

    # Majority vote
    unique, counts = np.unique(nearest_labels, return_counts=True)
    predicted = unique[np.argmax(counts)]

    return predicted, nearest_idx, nearest_dists

# Example: 2D classification
X_train = np.array([[2, 3], [5, 4], [9, 6], [4, 7], [8, 1]])
y_train = np.array(['Red', 'Red', 'Blue', 'Blue', 'Red'])
query = np.array([6, 5])

pred, idx, dists = knn_classify(X_train, y_train, query, k=3)
print(f"Query point: {query}")
print(f"3 nearest neighbors: {X_train[idx]}")
print(f"Distances: {np.round(dists, 2)}")
print(f"Labels: {y_train[idx]}")
print(f"Predicted class: {pred}")
```

```python
# Visualize KNN decision
colors = {'Red': '#C62828', 'Blue': '#4B2E83'}
fig = go.Figure()

for label in ['Red', 'Blue']:
    mask = y_train == label
    fig.add_trace(go.Scatter(
        x=X_train[mask, 0], y=X_train[mask, 1],
        mode='markers+text', name=f'Class {label}',
        marker=dict(size=14, color=colors[label]),
        text=[f'({x},{y})' for x, y in X_train[mask]],
        textposition='top center'))

fig.add_trace(go.Scatter(
    x=[query[0]], y=[query[1]], mode='markers',
    name='Query', marker=dict(size=18, color='#E8B923',
    symbol='diamond', line=dict(width=2, color='black'))))

# Draw circles to nearest neighbors
for i in idx:
    fig.add_shape(type="line",
        x0=query[0], y0=query[1],
        x1=X_train[i, 0], y1=X_train[i, 1],
        line=dict(color="gray", width=1, dash="dash"))

fig.update_layout(
    title=f'<b>KNN (k=3)</b> — Predicted: {pred}',
    xaxis_title='Feature X₁', yaxis_title='Feature X₂',
    template='plotly_white', height=450,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

```python
# Demonstrate curse of dimensionality on KNN
# Same 5 points, but in 1D, 2D, and 10D
np.random.seed(42)

results = []
for d in [1, 2, 5, 10, 50]:
    X = np.random.uniform(0, 10, (5, d))
    q = np.full(d, 5.0)
    dists = np.sqrt(np.sum((X - q)**2, axis=1))
    ratio = dists.max() / dists.min()
    results.append({'Dimensions': d,
                   'Min Distance': round(dists.min(), 2),
                   'Max Distance': round(dists.max(), 2),
                   'Max/Min Ratio': round(ratio, 2)})

df_curse = pd.DataFrame(results)
print("Distance concentration as dimensions grow:")
print(df_curse.to_string(index=False))
print("\nAs max/min → 1, KNN loses its discriminating power.")
```

<div style="border-left: 5px solid #4B2E83; background-color: #F9F7FC; padding: 20px 25px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; margin-top: 0;">Dot Product (Inner Product)</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;"><strong>u · v = Σ uᵢvᵢ = ‖u‖·‖v‖·cos θ</strong>. Measures alignment: positive = same direction, zero = perpendicular, negative = opposite. Used in cosine similarity, neural nets, linear regression.</p>
</div>

```python
# Dot product examples
u = np.array([2, 3, -1])
v = np.array([4, 0, 5])
dot = np.dot(u, v)
print(f"u = {u}, v = {v}")
print(f"u · v = (2)(4) + (3)(0) + (-1)(5) = {dot}")

# Orthogonality check
a = np.array([1, 2, -2])
b = np.array([4, -1, 1])
print(f"\na = {a}, b = {b}")
print(f"a · b = {np.dot(a, b)} → {'Orthogonal!' if np.dot(a, b) == 0 else 'Not orthogonal'}")
```

<h2 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; font-weight: 700; border-bottom: 3px solid #4B2E83; padding-bottom: 10px;">Norms (L1, L2, L∞)</h2>

```python
# Compare norms for v = (3, -4)
v = np.array([3, -4])
l1 = np.linalg.norm(v, 1)
l2 = np.linalg.norm(v, 2)
linf = np.linalg.norm(v, np.inf)
l4 = np.linalg.norm(v, 4)

print(f"v = {v}")
print(f"L1  (Manhattan):  ‖v‖₁ = |3| + |-4| = {l1}")
print(f"L2  (Euclidean):  ‖v‖₂ = √(9+16) = {l2}")
print(f"L∞  (Chebyshev):  ‖v‖∞ = max(|3|,|-4|) = {linf}")
print(f"L4  (general):    ‖v‖₄ = (81+256)^(1/4) = {l4:.3f}")
```

```python
# Visualize unit balls for different norms
theta = np.linspace(0, 2*np.pi, 500)

fig = go.Figure()
for p, name, color in [(1, 'L1 (Diamond)', '#E8B923'),
                        (2, 'L2 (Circle)', '#4B2E83'),
                        (4, 'L4', '#388E3C'),
                        (100, 'L∞ (Square)', '#C62828')]:
    x = np.sign(np.cos(theta)) * np.abs(np.cos(theta))**(2/p)
    y = np.sign(np.sin(theta)) * np.abs(np.sin(theta))**(2/p)
    fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name=name,
        line=dict(color=color, width=2.5)))

fig.update_layout(
    title='<b>Unit Balls</b> — Different norms define different geometries',
    xaxis=dict(scaleanchor='y', range=[-1.3, 1.3]),
    yaxis=dict(range=[-1.3, 1.3]),
    template='plotly_white', height=500, width=550,
    font=dict(family='Encode Sans, Arial, sans-serif'))
fig.show()
```

<h2 style="color: #4B2E83; font-family: 'Encode Sans', 'Arial', sans-serif; font-weight: 700; border-bottom: 3px solid #4B2E83; padding-bottom: 10px;">Cosine Similarity</h2>

```python
# Cosine similarity
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

u = np.array([3, 4])
v = np.array([4, 3])
cos_sim = cosine_similarity(u, v)
angle = np.degrees(np.arccos(cos_sim))

print(f"u = {u}, v = {v}")
print(f"cos(θ) = {cos_sim:.4f}")
print(f"θ ≈ {angle:.1f}° → {'Highly similar' if cos_sim > 0.9 else 'Moderate'}")
```

  <h3 style="color: #F57C00; margin-top: 0;">Exercise 1: Compute Euclidean Distance</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Given p = (1, 7, 3, 2) and q = (4, 3, 0, 4), compute d(p, q) = ‖p − q‖₂. Show each step.</p>

```python
# Exercise 1: Your code here
p = np.array([1, 7, 3, 2])
q = np.array([4, 3, 0, 4])

# Step 1: Subtract
diff = p - q
print(f'p - q = {diff}')

# Step 2: Square each
squared = diff**2
print(f'Squared: {squared}')

# Step 3: Sum and root
dist = np.sqrt(np.sum(squared))
print(f'd(p, q) = √{np.sum(squared)} = {dist:.4f}')
```

<div style="background-color: #FFF3E0; border-left: 4px solid #F57C00; padding: 15px 20px; margin: 20px 0; border-radius: 5px;">
  <h3 style="color: #F57C00; margin-top: 0;">Exercise 2: KNN Classification in 3D</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Given 6 training points in 3D, classify query q* = (3, 4, 2) using 3-NN. Compute all distances, identify 3 nearest, take majority vote.</p>
</div>

```python
# Exercise 2: 3D KNN
X = np.array([[1,2,1], [3,3,2], [6,5,5], [2,4,3], [7,6,4], [5,2,6]])
y = np.array(['Red','Red','Blue','Red','Blue','Blue'])
q = np.array([3, 4, 2])

# Compute distances
dists = np.sqrt(np.sum((X - q)**2, axis=1))
for i, (pt, label, d) in enumerate(zip(X, y, dists)):
    print(f"P{i+1} {pt} ({label}): d = {d:.2f}")

# 3-NN
idx = np.argsort(dists)[:3]
print(f"\n3 nearest: {['P'+str(i+1) for i in idx]}")
print(f"Labels: {y[idx]}")
unique, counts = np.unique(y[idx], return_counts=True)
print(f"Vote: {dict(zip(unique, counts))} → Predicted: {unique[np.argmax(counts)]}")
```

  <h3 style="color: #F57C00; margin-top: 0;">Exercise 3: Cosine Similarity Between Documents</h3>
  <p style="font-size: 1.05em; color: #555; line-height: 1.7;">Document 1 has term counts: (3, 0, 2, 1). Document 2 has: (1, 2, 0, 3). Compute cosine similarity and interpret.</p>

```python
# Exercise 3: Your code here
d1 = np.array([3, 0, 2, 1])
d2 = np.array([1, 2, 0, 3])

cos_sim = np.dot(d1, d2) / (np.linalg.norm(d1) * np.linalg.norm(d2))
print(f'cos(θ) = {cos_sim:.4f}')
```

```python
# ============================================================
#  LINEAR ALGEBRA EXPLORER DASHBOARD
# ============================================================
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display
import numpy as np

np.random.seed(42)

# ── Theme ──
C = {
    "bg": "#0F1117", "card": "#171A23", "grid": "#262B3A",
    "text": "#E8EAF0", "muted": "#8B90A0",
    "primary": "#4B2E83", "accent": "#E8B923",
    "green": "#22D3A5", "pink": "#E8608A",
    "palette": ["#4B2E83","#E8B923","#22D3A5","#E8608A","#4F8EF7","#F57C00"]
}
BASE = dict(paper_bgcolor=C["card"], plot_bgcolor=C["card"],
            font=dict(family="Inter, Segoe UI, sans-serif", color=C["text"], size=12),
            margin=dict(l=55, r=30, t=60, b=45))

def style(fig):
    fig.update_xaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    fig.update_yaxes(gridcolor=C["grid"], zerolinecolor=C["grid"], tickfont=dict(color=C["muted"]))
    return fig

figs = {}

# ── KPI: Norm comparison ──
v = np.array([3, -4])
norms = {"L1": np.linalg.norm(v,1), "L2": np.linalg.norm(v,2),
         "L∞": np.linalg.norm(v,np.inf), "L4": np.linalg.norm(v,4)}
f = make_subplots(rows=1, cols=4, specs=[[{"type":"indicator"}]*4], horizontal_spacing=0.05)
for i, (name, val) in enumerate(norms.items(), 1):
    f.add_trace(go.Indicator(mode="number", value=val,
        number=dict(valueformat=".3f", font=dict(size=36, color=C["text"])),
        title=dict(text=f"<span style='color:{C['muted']};font-size:14px'>‖v‖ {name}</span>")), 1, i)
f.update_layout(**BASE, height=160)
figs["kpi"] = f

# ── Unit balls ──
theta = np.linspace(0, 2*np.pi, 500)
f = go.Figure()
for p, name, col in [(1,'L1','#E8B923'),(2,'L2','#4B2E83'),(4,'L4','#22D3A5'),(100,'L∞','#E8608A')]:
    x = np.sign(np.cos(theta))*np.abs(np.cos(theta))**(2/p)
    y = np.sign(np.sin(theta))*np.abs(np.sin(theta))**(2/p)
    f.add_trace(go.Scatter(x=x, y=y, mode='lines', name=name, line=dict(color=col, width=2.5)))
f.update_layout(**BASE, height=420, title=dict(text="<b>Unit Balls — Lp Norms</b>", font=dict(size=16)),
               xaxis=dict(scaleanchor='y'))
style(f)
figs["balls"] = f

# ── Distance concentration ──
dims = [1, 2, 5, 10, 25, 50, 100, 250, 500, 1000]
ratios_list = []
for d in dims:
    from scipy.spatial.distance import pdist
    X = np.random.uniform(0, 1, (100, d))
    dd = pdist(X)
    ratios_list.append(dd.max() / dd.min())
f = go.Figure(go.Scatter(x=dims, y=ratios_list, mode='lines+markers',
    marker=dict(size=8, color=C["accent"]), line=dict(color=C["accent"], width=3)))
f.add_hline(y=1, line_dash="dash", line_color=C["pink"])
f.update_layout(**BASE, height=400, title=dict(text="<b>Curse of Dimensionality</b> — Distance Ratio Collapse", font=dict(size=16)),
               xaxis_title="Dimensions", yaxis_title="max/min ratio")
style(f)
figs["curse"] = f

# ── Cosine similarity heatmap ──
docs = np.array([[3,0,2,1],[1,2,0,3],[2,1,1,2],[0,3,1,0],[1,0,3,2]])
n = len(docs)
cos_mat = np.zeros((n,n))
for i in range(n):
    for j in range(n):
        cos_mat[i,j] = np.dot(docs[i],docs[j])/(np.linalg.norm(docs[i])*np.linalg.norm(docs[j]))
labels = [f"Doc {i+1}" for i in range(n)]
f = go.Figure(go.Heatmap(z=cos_mat, x=labels, y=labels,
    colorscale=[[0,"#171A23"],[0.5,"#4B2E83"],[1,"#E8B923"]],
    text=np.round(cos_mat, 2), texttemplate="%{text}", textfont=dict(size=12)))
f.update_layout(**BASE, height=420, title=dict(text="<b>Cosine Similarity Matrix</b>", font=dict(size=16)))
figs["cosine"] = f

# ── Assemble dashboard ──
def div(key, first=False):
    return figs[key].to_html(full_html=False,
        include_plotlyjs="cdn" if first else False,
        config={"displayModeBar": True, "displaylogo": False})

card = lambda inner: f"<div class='card'>{inner}</div>"
tabs = {
    "Norms": card(div("kpi", first=True)) + card(div("balls")),
    "Curse of Dim.": card(div("curse")),
    "Similarity": card(div("cosine")),
}
nav = "".join(f"<button class='tab-btn{' active' if i==0 else ''}' onclick=\"showTab(event,'tab{i}')\">"+name+"</button>" for i, name in enumerate(tabs))
panels = "".join(f"<div id='tab{i}' class='tab-panel' style='display:{'block' if i==0 else 'none'}'>{c}</div>" for i, c in enumerate(tabs.values()))

html = f"""<div id="dash"><style>
  #dash {{background:{C['bg']};border-radius:14px;padding:22px;font-family:Inter,'Segoe UI',sans-serif;}}
  #dash .dash-header {{display:flex;justify-content:space-between;align-items:baseline;margin-bottom:16px;}}
  #dash h1 {{color:{C['text']};font-size:21px;margin:0;}}
  #dash .sub {{color:{C['muted']};font-size:12px;}}
  #dash .tab-bar {{display:flex;gap:6px;border-bottom:1px solid {C['grid']};margin-bottom:18px;}}
  #dash .tab-btn {{background:none;border:none;color:{C['muted']};padding:10px 18px;font-size:13.5px;font-weight:600;cursor:pointer;border-bottom:2.5px solid transparent;}}
  #dash .tab-btn:hover {{color:{C['text']};}}
  #dash .tab-btn.active {{color:{C['accent']};border-bottom-color:{C['accent']};}}
  #dash .card {{background:{C['card']};border:1px solid {C['grid']};border-radius:12px;padding:6px;margin-bottom:16px;box-shadow:0 2px 10px rgba(0,0,0,.35);}}
</style>
<div class="dash-header"><h1>Linear Algebra Explorer</h1><span class="sub">BDATA 412 · Section 1</span></div>
<div class="tab-bar">{nav}</div>{panels}
<script>function showTab(evt,id){{const d=document.getElementById('dash');d.querySelectorAll('.tab-panel').forEach(p=>p.style.display='none');d.querySelectorAll('.tab-btn').forEach(b=>b.classList.remove('active'));document.getElementById(id).style.display='block';evt.currentTarget.classList.add('active');document.getElementById(id).querySelectorAll('.js-plotly-plot').forEach(g=>Plotly.Plots.resize(g));}}</script></div>"""
display(HTML(html))
print("Dashboard ready — switch tabs to explore Norms, Curse of Dimensionality, and Similarity.")
```
